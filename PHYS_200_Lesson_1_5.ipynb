<font size = 5> <b> 2.1 Recap Data Access in DataFrames</b> </font>

<p align="justify">
Let's take a brief look at what we saw in Lesson 1.4. Then we will expand on that. Let us begin by importing our celebrity heights data set.
</p>

```python
# Import the Pandas Library:

import pandas as pd

# Set the location of the CSV file (can be local or a website):

url = "https://raw.githubusercontent.com/dr-bankert-augustana/PHYS_200/refs/heads/main/Data/Module_1/celebrity-heights.csv"

# Turn the CSV data into a Pandas DataFrame:

celebrity_height_df = pd.read_csv(url)

# Display the DataFrame:

display(celebrity_height_df)
```

<b>Create a New DataFrame Based on Celebrity First Name</b>

Next, we create a new DataFrame object that only contains celebrities whose first name is "Matthew".

```python
matthew_only = celebrity_height_df[celebrity_height_df["firstname"] == "Matthew"]

matthew_only.head(10)
```

If we want to, we can restrict our DataFrame even more. Let's only grab the celebrities named "Matthew" that are at least six feet tall.

```python
six_foot_matthews = celebrity_height_df[(celebrity_height_df["firstname"] == "Matthew") & (celebrity_height_df["feet"] >= 6)]

six_foot_matthews.head(10)
```

<center>$\rule$</center>
<font size=5><b>2.2 Using the <code>query()</code> Function</b></font>

<p align="justify">
We have seen that we can effectively carve out slices of data from our DataFrames. However, the above method both looks and feels a bit clunky. Fortunately, there is a simpler, more natural feeling way to accomplish the same result using the <b>query()</b> function.
</p>

```python
six_foot_matthews2 = celebrity_height_df.query('firstname == "Matthew" and feet >= 6')

six_foot_matthews2.head(10)
```

<p align="justify">
Let's suppose that you aren't interested in all the celebrities named "Matthew", but are instead interested in a specific list of celebrities.

<br>

For example, lets use the cast of Star Trek: The Next Generation. We'll start by making a list of the names.
</p>

```python
list_of_celebrities = ["Patrick Stewart", "Brent Spiner", "Jonathan Frakes", "Gates McFadden", "Levar Burton", "Michael Dorn", "Marina Sirtis"]
```

Next, we'll look at our DataFrame and pick out only celebrities whose full name is in that list using the <b>isin()</b> function.

```python
star_trek_tng_cast = celebrity_height_df[celebrity_height_df["fullname"].isin(list_of_celebrities)]

display(star_trek_tng_cast)
```

Again, we can also accomplish this task using the <code>query()</code> function.

```python
star_trek_tng_cast2 = celebrity_height_df.query("fullname in @list_of_celebrities")

display(star_trek_tng_cast2)
```

As before, the <b>query()</b> makes it easy and natural to ask multiple questions of the data at once.

Let's get only those celebrities on our list that are over six feet tall.

```python
six_foot_star_trek_tng_cast = celebrity_height_df.query("fullname in @list_of_celebrities and feet >= 6")

display(six_foot_star_trek_tng_cast)
```

<font size = 5> <b> 3.1 Iris Dataset </b> </font>

<p align="justify">
To begin our exploration of plotting with Pandas, we will be looking at a dataset on Irises from the Python library sklearn. The Iris dataset is a classic and widely used dataset in machine learning, particularly for demonstrating classification algorithms. It is readily available within the sklearn.datasets module in scikit-learn.
</p>

<br>

<b>Key characteristics of the Iris dataset:</b>

  * <b>Origin:</b><p align="justify">
    Compiled by Ronald Fisher in 1936, this dataset is often considered the "hello world" of machine learning for supervised classification problems.
    </p>

  * <b>Content:</b><p align="justify">
    It contains measurements of three different species of Iris flowers: Iris setosa, Iris versicolor, and Iris virginica.
    </p>

  * <b>Features:</b><p align="justify">
    Each flower sample is described by four numerical features:
      * Sepal length (cm)
      * Sepal width (cm)
      * Petal length (cm)
      * Petal width (cm)
    </p>

  * <b>Target:</b><p align="justify">
    The target variable represents the species of the Iris flower, encoded as numerical values (0 for setosa, 1 for versicolor, and 2 for virginica).
    </p>

  * <b>Size:</b><p align="justify">
    The dataset consists of 150 samples, with 50 samples for each of the three species, making it a balanced dataset.
    </p>

```python
# Load Datasets from Sklearn:

import seaborn as sns

# Import Iris (flower) Species Dataset as a DataFrmae:

iris_df = sns.load_dataset("iris")

# Get Basic Info With DataFrame.info()

print("Iris DataFrame Basic Info:\n")

display(iris_df.info())

print('\n\n')

# Get Statistical Info on Numeric Columns with DataFrame.describe()

print("Iris DataFrame Statistical Info:\n")

display(iris_df.describe())

print('\n\n')

# Get a Look at the DataFrame with DataFrame.head()

print("Iris DataFrame Head:\n")

display(iris_df.head())
```

<font size=5><b>3.2 The <code>DataFrame.plot()</code> Function</b></font>

<p align="justify">
[`DataFrame.plot()`](https://pandas.pydata.org/pandas-docs/stable/reference/api/pandas.DataFrame.plot.html) is an all-purpose tool for plotting data in DataFrames using Matplotlib under the hood.
</p>

+ See the documentation for important parameters: `x`, `y`, `kind`, etc.

<br>

<p align="justify">
There are many different kinds of plots that you can create using this function. Let's explore a few of the more common types. You set the type of plot using the <code>kind = 'type'</code> keyword in your call to the plotting function.
</p>

<br>

<b>Histogram: (kind = 'hist'):</b>

<p align="justify">
A histogram plot is a type of bar chart that represents the distribution of a dataset. It shows how often (frequency) different ranges (or "bins") of values occur in the data.
</p>

Key Features of a Histogram:

  * X-axis (horizontal): Represents intervals (or "bins") of the data.

  * Y-axis (vertical): Represents the frequency (number of data points) within each bin.

  * Bars: Each bar represents the number of observations within a specific range.

<br>

<b>Code Example:</b>

In the following code cell, we use a histogram plot to show the frequency of sepal width and sepal length measurements.

```python
display(iris_df.plot(kind = 'hist', y = ['sepal_length', 'sepal_width']))
```

We can also use Matplotlib to help customize our graphs.

```python
# Import Matplotlib library:

import matplotlib.pyplot as plt

# Create the histogram plot

ax = iris_df.plot(
    kind='hist',
    y = ['sepal_length', 'sepal_width'],
    bins=20,
    alpha=0.6,                     # Make overlapping histograms transparent
    color=['teal', 'darkorange'],  # Custom colors
    edgecolor='black',
    figsize=(8, 5)
)

# Use Matplotlib to customize the plot:

# Add titles and labels:

ax.set_title("Distribution of Sepal Length and Width", fontsize=14)
ax.set_xlabel("Measurement (cm)", fontsize=12)
ax.set_ylabel("Frequency", fontsize=12)

# Add grid and legend:

plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(title="Features")

plt.tight_layout()
plt.show()
```

<b>Bar Plot: (kind = 'bar'):</b>

<p align="justify">
A bar plot (or bar chart) is a type of graph used to visually represent categorical data using rectangular bars. The height (or length) of each bar reflects the value or frequency of the corresponding category.
</p>

Key Features of a Bar Plot:

  * X-axis: Represents categories (e.g., fruit types, countries, brands).

  * Y-axis: Represents numerical values (e.g., count, average, sales).

  * Bars: Each bar's height (vertical) or width (horizontal) shows the value of that category.

<b>Code Example:</b>

In the following code cell, we plot the average petal length by Species.

```python
# Group by species and calculate mean petal length:

mean_petal_length = iris_df.groupby("species")["petal_length"].mean()

# Plot bar chart:

mean_petal_length.plot(kind = 'bar', color = ['lightblue', 'lightgreen', 'salmon'])

# Use Matplotlib to customize the plot:

plt.title("Average Petal Length by Iris Species")
plt.xlabel("Species")
plt.ylabel("Average Petal Length (cm)")
plt.xticks(rotation = 0)
plt.grid(axis = 'y', linestyle = '--', alpha = 0.7)
plt.tight_layout()
plt.show()
```

<b>Scatter Plot: (kind = 'scatter'):</b>

<p align="justify">
A scatter plot is a type of plot or graph that displays individual data points on a two-dimensional plane using Cartesian coordinates. It is used to observe the relationship (or correlation) between two numerical variables.
</p>

Key Features of a Scatter Plot:

  * X-axis: Represents one variable.

  * Y-axis: Represents another variable.

  * Each point: Represents a single observation (x, y pair).

```python
display(iris_df.plot(kind = 'scatter',x = 'sepal length (cm)',y = 'sepal width (cm)'))
```

<b>Line Plot: (kind = 'line'):</b>

<p align="justify">
A line graph (or line plot) is a type of chart used to display data points connected by straight lines. It's especially useful for showing trends over time or continuous data.
</p>

Key Features of a Line Plot:

  * X-axis (horizontal): Typically shows time or another continuous variable (e.g. days, months, distance).

  * Y-axis (vertical): Shows the value of the variable being measured.

  * Data points: Represent individual measurements.

  * Lines: Connect the data points to show changes or trends over the range.

<br>

<b>Code Example:</b>

In the following code cell, we visualize how the average measurements of flower parts differ across species.

```python
# Group by species and calculate mean for each measurement

mean_values = iris_df.groupby("species").mean(numeric_only = True)

# Transpose so species become columns for easy line plotting

mean_values.T.plot(kind = 'line', marker = 'o')

# Use Matplotlib to customize the plot:

plt.title("Average Iris Measurements by Species")
plt.xlabel("Measurement Type")
plt.ylabel("Mean Value (cm)")
plt.grid(True)
plt.legend(title="Species")
plt.tight_layout()
plt.show()
```

<font size=5><b>3.4 Correlation Matrix and Heat Maps</b></font>

<p align="justify">
A correlation matrix is a table that shows the correlation coefficients between many variables in a dataset. It helps you understand the strength and direction of relationships between pairs of numerical features.
</p>

<br>

It's a square matrix, where:

  * Rows and columns represent the same variables.

  * Each cell shows the correlation coefficient between the variables of that row and column.

<br>

Correlation Coefficient (usually Pearson's r):

  * Ranges from -1 to 1

  * 1	Perfect positive correlation

  * 0	No correlation

  * -1	Perfect negative correlation

<br>

<b>Example:</b>

Imagine you're managing an ice cream truck, and you track two things daily:

  1. The temperature outside.

  2. The number of ice creams sold.

There will likely be a strong positive correlation. As temperature goes up, so do sales. The correlation value (Pearson's r) would be close to +1.

<br>

<b>Other Examples:</b>

  * $\approx$ +1:	More temperature = more water bottle sales
  * $\approx$ 0:	No correlation	Temperature vs. shoe size (unrelated)
  * $\approx$ -1:	More clouds = fewer sunglasses sold

<br>

<b>Heatmap:</b>

<p align="justify">
A heat map is a visual representation of data where values are shown as colors. It's often used to display 3D data in a 2D format, where:

  * One set of colors indicate larger values.

  * A seperate set of colors indicate lesser values.

Key Features of Heat Map:

  * X and Y axes: positional data (real or phase space).

  * Cell color: Represents the value of the property at the x,y location.

  * Color scale (colormap): Helps interpret what the colors mean (e.g., blue = negative, red = positive).

<br>

<b>Code Example:</b>

In the following code cell, we create a correlation matrix for our Iris DataFrame. We also display the matrix as a heatmap.

```python
# Compute correlation matrix (numerical columns only):

corr_matrix = iris_df.corr(numeric_only = True)

print("Correlation Matrix:")
display(corr_matrix)

print('\n\n')

# Plot the correlation matrix as a heatmap:

sns.heatmap(corr_matrix, annot = True, cmap = "coolwarm", center = 0)
plt.title("Correlation Matrix of Iris Features")
plt.show()
```
