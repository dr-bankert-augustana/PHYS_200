<font size=5><b>2.3 Importing Libraries in Python</b></font>

Before we can use any external library in Python, we need to <b>import</b> it using the `import` statement.

<br>

For example, to use NumPy, we write:

```python
import numpy as np
```

<br>

<p align="justify">
This tells Python to load the definitions in the NumPy library and make it available under the <a href = #Definition-Alias></b>alias</b></a> `np`. An alias is like a nickname. It lets us refer to a library with a shorter, more convenient name so that we don't need to write out the full name of the library when we want to use its contents.
</p>

<br>

Using an alias, we can write:

```python
print(np.sqrt(4))
```

<br>

Instead of using the full library name, as we would have to do if we didn't use an alias:

```python
print(numpy.sqrt(4))
```

<br>

<p align="justify">
This might not seem like such a big deal at first as `numpy` and `np` are close to the same length, but <b>I promise you</b> that you'd rather write `plt` than `matplotlib.pyplot` over and over again in your code.

<p align="justify">
The first library we'll introduce is called <a href = #Definition-Numpy></b>NumPy</b></a>. It stands for "Numerical Python". NumPy is one of the foundational tools for scientific computing in Python. A powerful toolbox of mathematical functions (many of which are either missing or less efficient in standard Python), it provides efficient array structures (like vectors and matrices) and supports a wide range of mathematical operations on these arrays, all optimized for speed and performance.

<b>Code Example:</b>

In the following Code cell, we import NumPy and and showcase some of its basic mathematical functions.

````python
# Import the NumPy library

import numpy as np

# Now that the library is imported, we can use its functions.

# Let's explore some basic NumPy math functions. (We can use round() to truncate the output of the numbers.)

# Square root of 2:

print("Square root of 2 (no rounding):", np.sqrt(2))
print("Square root of 2 (rounded to 3 decimal places):", round(np.sqrt(2), 3))
print()

# Cosine of π:

print("Cosine of π:", np.cos(np.pi))
print()

# Exponential (e^x):

print("Exponential (base e) of 1:", round(np.exp(1), 3))
print()

# Natural logarithm of 10:

print("Natural log of 10:", round(np.log(10), 3))
print()

# Sine of π/2:

print("Sine of π/2:", np.sin(np.pi / 2))
````

<font size=5><b>3.2 NumPy Arrays vs Python Lists</b></font>

<p align="justify">
As we saw in our previous lessons, in standard Python, the arrays that we use are called "lists"; we use these arrays to store collections of data. While native Python lists are flexible and easy to use, they're not optimized for numerical operations. For example, multiplying a list by a number repeats the list. It doesn't multiply the values inside it.
</p>

<br>

<p align="justify">
NumPy arrays, on the other hand, are designed specifically for numerical data. They support element-wise operations, are more memory efficient, and can be much faster than lists, especially for large datasets.
</p>

<br>

<b>Code Example:</b>

Let's look at some examples to compare how Python lists and NumPy arrays behave.

````python
# Import the NumPy library:

import numpy as np

##=============================================================================================##
## Create a Python list and a NumPy array:                                                     ##
##=============================================================================================##

py_list = [1, 2, 3, 4, 5]

np_array = np.array([1, 2, 3, 4, 5])

##=============================================================================================##
## Print both objects:                                                                         ##
##=============================================================================================##

print("Python list:", py_list)
print()
print("NumPy array:", np_array)
print()

##=============================================================================================##
## Multiply both objects by a scalar:                                                          ##
##=============================================================================================##

# Multiply Python list by 2 (Repeats the list)

print("Python list * 2:", py_list * 2)
print()

# Multiply NumPy array by 2 (Multiplies each element by 2)

print("NumPy array * 2:", np_array * 2)
print()

##=============================================================================================##
## Add a scalar value to both objects:                                                         ##
##=============================================================================================##


# Add a scalar value to the Python list (creates an error):
# Note: The (try) and (except) commands allow us to have an error without crashing the program.

try:

  print(py_list + 2)

except TypeError as e:

  print("TypeError:", e)
  print()

# Add a scalar value to the Numpy array (adds 2 to each element):

print("NumPy array + 2:", np_array + 2)
````

<font size=5><b>2.3 Native Python vs. NumPy: Loops vs. Vectorization</b></font>

Let’s say we want to compute the square root of a list of numbers.

Using **native Python**, we’d typically use a `for` loop to apply the square root to each value. But with **NumPy**, we can apply the operation to the entire array at once — this is called **vectorization**.

Not only is the NumPy approach **shorter and more readable**, it’s also **much faster**, especially for large datasets.


````python
import math
import numpy as np

# A list of numbers from 1 to 10
numbers = list(range(1, 11))

# Native Python using a for loop
sqrt_python = []
for x in numbers:
    sqrt_python.append(math.sqrt(x))

# NumPy vectorized version
numbers_np = np.array(numbers)
sqrt_numpy = np.sqrt(numbers_np)

# Output both results
print("Native Python (loop):", sqrt_python)
print("NumPy (vectorized):  ", sqrt_numpy.tolist())  # Convert to list for comparison
````

🚀 **Key Takeaway**

Vectorized operations with NumPy are:

- **Cleaner**: No need to write loops.
- **Faster**: NumPy operations run in compiled C code under the hood.
- **More Powerful**: You can apply functions to entire arrays or matrices with a single line of code.

This is one of the main reasons NumPy is essential for scientific and numerical computing in Python.
