<font size=5><b>2.3 Importing Libraries in Python</b></font>

Before we can use any external library in Python, we need to <b>import</b> it using the `import` statement.

<br>

For example, to use NumPy, we write:

```python
import numpy as np
```

<br>

<p align="justify">
This tells Python to load the definitions in the NumPy library and make it available under the <a href = #Definition-Alias></b>alias</b></a> <code>np</code>. An alias is like a nickname. It lets us refer to a library with a shorter, more convenient name so that we don't need to write out the full name of the library when we want to use its contents.
</p>

<br>

Using an alias, we can write:

```python
print(np.sqrt(4))
```

<br>

Instead of using the full library name, as we would have to do if we didn't use an alias:

```python
print(numpy.sqrt(4))
```

<br>

<p align="justify">
This might not seem like such a big deal at first as `numpy` and `np` are close to the same length, but <b>I promise you</b> that you'd rather write `plt` than `matplotlib.pyplot` over and over again in your code.
</p>

<font size = 5> <b> 3.1 Introducing NumPy </b> </font>

<p align="justify">
The first library we'll introduce is called <a href = #Definition-NumPy></b>NumPy</b></a> (pronounced "Numb-Pie" and short for <b>Numerical Python</b>) is the foundational library for numerical computing in Python. A powerful toolbox of mathematical functions (many of which are either missing or less efficient in standard Python), it provides efficient array structures (like vectors and matrices) and supports a wide range of mathematical operations on these arrays, all optimized for speed and performance.
</p>

<br>

<b>Code Example:</b>

In the following code cell, we import NumPy and and showcase some of its basic mathematical functions.

````python
# Import the NumPy library
# (Note: If you already imported NumPy earlier, you don't need to import it again.
#  We do it here for illustrative purposes.)

import numpy as np

# Now that the library is imported, we can use its functions.

# Let's explore some basic NumPy math functions. (We can use round() to truncate the output of the numbers.)

# Square root of 2:

print("Square root of 2 (no rounding):", np.sqrt(2))
print("Square root of 2 (rounded to 3 decimal places):", round(np.sqrt(2), 3))
print()

# Cosine of π:

print("Cosine of π:", np.cos(np.pi))
print()

# Exponential (e^x):

print("Exponential (base e) of 1:", round(np.exp(1), 3))
print()

# Natural logarithm of 10:

print("Natural log of 10:", round(np.log(10), 3))
print()

# Sine of π/2:

print("Sine of π/2:", np.sin(np.pi / 2))
````

<center>$\rule$</center>
<font size=5><b>3.2 NumPy Arrays vs Python Lists</b></font>

<p align="justify">
As we saw in our previous lessons, in standard Python, the arrays that we use are called "lists"; we use these arrays to store collections of data. While native Python lists are flexible and easy to use, they're not optimized for numerical operations. For example, multiplying a list by a number repeats the list. It doesn't multiply the values inside it.
</p>

<br>

<p align="justify">
NumPy arrays, on the other hand, are designed specifically for numerical data.
</p>

  * NumPy arrays are <b>faster</b> and <b>more memory-efficient</b> (especially for large datasets).
  * NumPy provides <b>built-in vectorized operations</b> (less need to loop over internal elements).
  * Numpy arrays provide a more natural structure for <b>multi-dimensional arrays</b>, <b>linear algebra</b>, <b>statistics</b>, and more.

<br>

<b>Code Example:</b>

In the following code cell, we look at some examples to compare how Python lists and NumPy arrays behave.

````python
# Import the NumPy library:
# (Note: If you already imported NumPy earlier, you don't need to import it again.
#  We do it here for illustrative purposes.)

import numpy as np

##=============================================================================================##
## Create a Python list and a NumPy array:                                                     ##
##=============================================================================================##

py_list = [1, 2, 3, 4, 5]

np_array = np.array([1, 2, 3, 4, 5])

##=============================================================================================##
## Print both objects:                                                                         ##
##=============================================================================================##

print("Python list:", py_list)
print()
print("NumPy array:", np_array)
print('\n\n')

##=============================================================================================##
## Multiply both objects by a scalar:                                                          ##
##=============================================================================================##

# Multiply Python list by 2 (Repeats the list)

print("Python list * 2:", py_list * 2)
print()

# Multiply NumPy array by 2 (Multiplies each element by 2)

print("NumPy array * 2:", np_array * 2)
print('\n\n')

##=============================================================================================##
## Add a scalar value to both objects:                                                         ##
##=============================================================================================##

# Add a scalar value to the Python list (creates an error):
# Note: The (try) and (except) commands allow us to have an error without crashing the program.

try:

  print(py_list + 2)

except TypeError as e:

  print("TypeError:", e)
  print()

# Add a scalar value to the Numpy array (adds 2 to each element):

print("NumPy array + 2:", np_array + 2)
print('\n\n')

##=============================================================================================##
## Add two arrays:                                                                             ##
##=============================================================================================##

# Create the Python lists:

list_1 = [1, 2, 3]
list_2 = [4, 5, 6]

# Create the NumPy arrays:

array_1 = np.array(list_1)
array_2 = np.array(list_2)

# Add the lists together:

list_sum = list_1 + list_2
print("Python list sum:", list_sum)
print()

# Add the arrays together:

array_sum = array_1 + array_2
print("NumPy array sum:", array_sum)
````

<font size=5><b>3.3 Native Python vs. NumPy: Loops vs. Vectorization</b></font>

<p align="justify">
Let's say that we want to compute the squares of a list of numbers. Using <b>native Python</b>, we typically need to use a <code>for</code> loop to apply the square to each value. But with <b>NumPy</b>, we can apply the operation to the entire array at once. This is called <b>vectorization</b>. Not only is the NumPy approach <b>shorter and more readable</b>, it's also <b>much faster</b>, especially for large datasets.
</p>

<br>

<b>Code Example:</b>

In the following code cell, we create arrays of values and then compute the squares of the elements of those arrays. We compare the method of doing so in native Python and in Numpy.

````python
# Import the NumPy library:

import numpy as np

##=============================================================================================##
## Create the arrays:                                                                          ##
##=============================================================================================##

# Create a native Python list of numbers from 1 to 10:

numbers_list = list(range(1, 11))

# Create a NumPy array of numbers from 1 to 10:

numbers_np = np.array(numbers_list)

##=============================================================================================##
## Compute the squares of the array elements:                                                  ##
##=============================================================================================##

# In native Python using a for loop:

squared_list = []

for number in numbers_list:

  squared_list.append(number**2)

# In NumPy using vectorized operations:

squared_array = np.pow(numbers_np, 2)

##=============================================================================================##
## Output the results of the operations:                                                       ##
##=============================================================================================##

print("Native Python (loop):", squared_list)
print()
print("NumPy (vectorized):  ", squared_array.tolist())  # Convert to list for ease of comparison
````

<b>Key Takeaway</b>

Vectorized operations with NumPy are:

  * <b>Cleaner</b>: No need to write loops.
  * <b>Faster</b>: NumPy operations run in compiled C code under the hood.
  * <b>More Powerful</b>: You can apply functions to entire arrays or matrices with a single line of code.

<br>

This is one of the main reasons NumPy is essential for scientific and numerical computing in Python.

<font size=5><b>5.1 Introducing Matplotlib</b></font>

<p align="justify">
When working with data, especially in science and engineering, visualization is essential. This is where <a href = #Definition-Matplotlib></b>Matplotlib</b></a> comes in.
</p>

<br>

<p align="justify">
Matplotlib is a powerful plotting library for Python. It allows you to create a wide range of plots, from simple line graphs to complex 2D and 3D visualizations. It's especially useful in physics and computational science for visualizing functions, simulation results, or experimental data.
</p>

</br>

The most commonly used part of Matplotlib is the pyplot module, which is typically imported as: <code>import matplotlib.pyplot as plt</code>.
</p>

<br>

<b>Code Example</b>

In the following code cell, we go over the basic structure of using Matplotlib.pyplot to create a graph. In general, the steps are:

  * Define your data (e.g. x and y values).

  * Use plotting functions like <code>plt.plot()</code>, <code>plt.scatter()</code>, etc.

  * Customize your plot by adding titles, labels, legends, etc.

  * Use <code>plt.show()</code> to display the plot or <code>plt.savefig()</code> to save a copy of your plot.

````python
# Import the libraries

import numpy as np
import matplotlib.pyplot as plt

##=============================================================================================##
## 1. Create data for the plot (or import it from somewhere):                                  ##
##=============================================================================================##

# Create an array of x-values from -10 to 10 (100 data points):

x = np.linspace(-10, 10, 100)

# Compute the corresponding y-values for a simple function (y = x^2):

y = x**2

##=============================================================================================##
## 2. Create the plot:                                                                         ##
##=============================================================================================##

# Plot the function with a label:

plt.plot(x, y, label = "y = x²", color = "blue")

# Add title and axis labels:

plt.title("Plot of y = x²")
plt.xlabel("x")
plt.ylabel("y")

# Show a legend

plt.legend()

# Display the plot

plt.show()

##=============================================================================================##
## 3. (Optional) Save a version of your plot to a file:                                        ##
##=============================================================================================##

plt.savefig("my_plot.png")
````

<center>$\rule$</center>
<font size=5><b>5.2 Customizing Plots</b></font>

Matplotlib gives you full control over the appearance of your plots. Here are a few things you can easily customize:

  * Line styles and colors

  * Markers (dots on data points)

  * Axis limits

  * Multiple plots on the same axes

  * Grid lines and annotations

<br>

<b>Code Example:</b>

In the following code cell, we try plotting multiple functions on the same graph.

````python
##=============================================================================================##
## Plot multiple functions: y = x, y = x², y = x³:                                             ##
##=============================================================================================##

# Create our x-axis data (an array of 100 data points from -5 to 5:

x = np.linspace(-5, 5, 100)

# Compute the corresponding y-values for our functions:

y1 = x

y2 = x**2

y3 = x**3

# Create the first plot:

plt.plot(x, y1, label = "y = x", linestyle = "--", color = "green")

# Create the second plot:

plt.plot(x, y2, label = "y = x²", linestyle = "-", color = "blue")

# Create the third plot:

plt.plot(x, y3, label = "y = x³", linestyle = ":", color = "red")

# Add title and axis labels:

plt.title("Comparison of Functions")
plt.xlabel("x-axis")
plt.ylabel("y-axis")

#Add a legend:

plt.legend()

# Add a grid for better readability:

plt.grid(True)

# Display the graph:

plt.show()

# Save a copy of the graph to a file:

plt.savefig("my_plot_2.png")
````

<font size=5><b>6.1 Introducing Seaborn</b></font>

<p align="justify">
While Matplotlib is powerful, it often requires a lot of manual setup to make plots look clean and professional. <a href = #Definition-Seaborn></b>Seaborn</b></a> is a library that builds on Matplotlib the same way that SciPy builds on Numpy.
</p>

<br>

<p align="justify">
Seaborn is a high-level data visualization library built on top of Matplotlib. It makes it easier to create attractive, informative, and statistically meaningful plots, especially when working with large datasets.
</p>

<br>

With Seaborn, you can:

  * Create beautiful plots with less code

  * Visualize relationships between variables

  * Work directly with pandas DataFrames (we're introduced to these in the next lesson).

  * Generate plots with built-in statistical features (like regression lines, boxplots, and violin plots)

<br>

<b>Code Example:</b>

To get started, we usually import Seaborn like this: <code>import seaborn as sns</code>. Let's begin with a basic example using a built-in dataset.

````python
# Import necessary libraries

import seaborn as sns
import matplotlib.pyplot as plt

##=============================================================================================##
## 1. Load a sample dataset:                                                                   ##
##=============================================================================================##

# Seaborn includes several built-in datasets for testing and learning.
# This is data about restaurant bills and tips.

tips = sns.load_dataset("tips")

##=============================================================================================##
## 2. Create a simple scatter plot with Seaborn:                                               ##
##=============================================================================================##

# Plot total bill vs. tip, colored by gender:

sns.scatterplot(data = tips, x = "total_bill", y = "tip", hue = "sex")

# Add a title:

plt.title("Total Bill vs. Tip (colored by sex)")

# Show the plot:

plt.show()

# Save a copy of the plot:

plt.savefig("my_plot_3.png")
````

<center>$\rule$</center>

Seaborn includes many useful plot types, such as:

  * <b>sns.histplot():</b> for histograms

  * <b>sns.boxplot():</b> for distributions and outliers

  * <b>sns.violinplot():</b> for visualizing distribution and density

  * <b>sns.barplot():</b> for comparing group averages

  * <b>sns.lineplot():</b> for trend lines

  * <b>sns.heatmap():</b> for visualizing matrices

<br>

<b>Code Example</b>

Let's try another example: a histplot to show how tips vary by day of the week.

````python
# Create a barplot of tips by day

sns.barplot(x = "day", y = "total_bill", hue = "day", data = tips, palette = "viridis", legend = False)

# Add Title and labels:

plt.title("Distribution of Tips by Day")
plt.xlabel("Day of the Week")
plt.ylabel("Tip Amount ($)")

# Display the plot:

plt.show()

# Save a copy of the plot:

plt.savefig("my_plot_4.png")
````
