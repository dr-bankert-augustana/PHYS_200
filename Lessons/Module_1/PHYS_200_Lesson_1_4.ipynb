<font size=5><b>3.1 Creating DataFrames From Dictionaries</b></font>

<p align="justify">
One of the most straightforward ways to create a DataFrame in Pandas is by using a Python dictionary. In this approach, each key in the dictionary represents a column name, and the corresponding value is a list (or array) containing the data for that column. Pandas automatically aligns these values by row, as long as the lists are of equal length.
</p>

<br>

<p align="justify">
This method is especially useful when you already have structured data in Python (for example, experimental measurements or simulation results stored in variables) and you want to organize them into a tabular format for further analysis.
</p>

<br>

<b>Code Example:</b>

In the following code cell, we create a simple DataFrame from some synthetic experimental data.

```python
# Import the Pandas Library:

import pandas as pd

# Create a dictionary of data:

data = {
    "Time (s)": [0, 1, 2, 3, 4],
    "Position (m)": [0, 2.5, 10, 22.5, 40],
    "Velocity (m/s)": [0, 5, 10, 15, 20]
}

# Convert the dictionary into a DataFrame:

df = pd.DataFrame(data)

# Display the DataFrame:

display(df)
```

<b>Understanding the Output</b>

<p align="justify">
This printed table looks very similar to what you'd see in a spreadsheet like Excel or Google Sheets:
</p>

  * <b>Columns:</b> Each column corresponds to one of the keys from the dictionary. In this case, we have three columns: <code>Time (s)</code>, <code>Position (m)</code>, and <code>Velocity (m/s)</code>. These serve the same purpose as column headers in a spreadsheet, labeling the type of data in each column.

  * <b>Rows:</b> Each row contains a single set of related data points; one time value, one position, and one velocity. These rows represent observations or measurements taken at different points in time.

  * <b>Index column (on the far left):</b> By default, Pandas automatically adds an index column starting from 0. This index uniquely identifies each row, similar to row numbers in a spreadsheet. While it doesn't carry physical meaning in this example, it becomes very useful when accessing or referencing specific rows programmatically.

<br>

<p align="justify">
So, in essence, the Pandas DataFrame behaves much like a digital spreadsheet, but with the added power of Python. As we see in this lesson, you can apply mathematical operations, filter rows, compute statistics, and generate plots all from this structured format, making it a key tool in your computational physics toolkit.
</p>

<font size=5><b>5.1 Bracket Notation []</b></font>

<p align="justify">
Once we have loaded and explored our DataFrame, the next step is accessing specific columns or subsets of data. One of the most common and flexible ways to do this is using bracket selection, which looks like: <code>DataFrame[selector]</code>
</p>

<br>

<p align="justify">
Depending on what we put inside the brackets, we can select a column, multiple columns, or filter rows using logical conditions.
</p>

<center>$\rule$</center>
<font size=5><b>5.2 Selecting a Single Column</b></font>

<p align="justify">
If we pass a string (column name) inside the brackets, Pandas returns a Series (a one-dimensional view of that column). This works like selecting a column in a spreadsheet by its header. The result behaves like a list but with labels (index) attached.
</p>

<br>

><b>Note:</b><p align="justify">
We can also call up a column by using: <code>DF_Name.Column_Name</code>
</p>

<br>

<b>Code Example:</b>

```python
# Select the 'firstname' column of the celebrity heights data:

first_names = celebrity_height_df['firstname']

# Display the firstnames data:

display(first_names)

# Select the 'lastname' column of the celebrity heights data:

last_names = celebrity_height_df.lastname

# Display the lastnames data:

display(last_names)
```

<center>$\rule$</center>
<font size=5><b>5.3 Selecting a Multiple Columns</b></font>

<p align="justify">
To select more than one column, pass in a list of column names inside the brackets: <code>subset_df = df[[column_1, column_2]]</code>. Now you get a new DataFrame with just those columns. Similar to hiding other columns in a spreadsheet.
</p>

<br>

<b>Code Example:</b>

```python
# Select the 'firstname' and 'lastname' columns of the celebrity heights data:

names = celebrity_height_df[['firstname', 'lastname']]

# Display the subset data:

display(names)
```

<center>$\rule$</center>
<font size=5><b>5.4 Selecting Rows with Logical Conditions</b></font>

<p align="justify">
You can also use bracket selection with boolean expressions to filter rows. This is incredibly powerful for working with physical measurements or simulation outputs. This is called using a <b>Boolean Mask</b>. Example: <code>subset = df[boolean_mask]</code>. This filters the DataFrame and only returns rows that meet the condition, just like using filters in Excel or Google Sheets.
</p>

><b>Note:</b><p align="justify">
You can build more complex filters by combining multiple conditions.
</p>

<br>

<b>Code Example:</b>

```python
# Create a subset of people who are over 2.2 meters tall:

tall_people = celebrity_height_df[celebrity_height_df['meters'] > 2.2]

# Display the subset data:

print("Tall People:")
display(tall_people)
print()

# Create a subset of people who are between 1.6 and 1.7 meters tall:

medium_tall_people = celebrity_height_df[(celebrity_height_df['meters'] > 1.6) & (celebrity_height_df['meters'] < 1.7)]

# Display the subset data:

print("Medium Tall People:")
display(medium_tall_people)
print()

# Create a subset of people who are shorter than 1.1 meters and taller than 2.2  meters tall:

short_and_tall_people = celebrity_height_df[(celebrity_height_df['meters'] < 1.1) | (celebrity_height_df['meters'] > 2.2)]

# Display the subset data:

print("Short and Tall People:")
display(short_and_tall_people)
```

<font size=5><b>5.5 <code>.loc()</code> and <code>.iloc()</code></b></font>

<p align="justify">
Pandas provides two powerful accessors:
</p>

  * .loc[]: selects by label

  * .iloc[]: selects by integer position

<br>

<p align="justify">
These are much more versatile than [] alone, especially when selecting specific rows and columns at the same time or when doing row slicing.
</p>

  * .loc[]: Selection by Label

    * Syntax: <code>df.loc[row_selector, column_selector]</code>

  * Selecting Rows by Index Label

    * Syntax: <code>df.loc[0]  # Selects the row with index label 0</code>

  * Row Slicing by Label

    * Syntax: <code>df.loc[0:2]</code>

    ><b>Note:</b> <code>.loc[]</code> slicing includes the stop value (unlike standard Python slicing).

  * Selecting Specific Rows and Columns by Label

    * Syntax: <code>df.loc[0:2, ["Time (s)", "Velocity (m/s)"]]</code>

    * This returns rows 0 through 2, and only the "Time" and "Velocity" columns.

  * .iloc[]: Selection by Integer Position

    * Syntax: df.iloc[row_index, column_index]

    * Use .iloc[] when you want to slice or select based on numerical index positions, just like Python lists.

  * Selecting a Row by Integer Index

    * df.iloc[0]  # First row

  * Row Slicing by Position

    * df.iloc[0:3]  # Rows 0 through 2 (excludes stop index, like Python slicing)

  * Selecting Specific Rows and Columns by Position

    * df.iloc[0:3, 0:2]  # First 3 rows, first 2 columns

## Group Celebrities Based on First Name

Lets say that we want to find out information about a specific subset within our DataFrame. We can do so by grouping our entries based on some criteria.

For example, lets group all the celebrities in our DataFrame based on their first names. The result is a <b>DataFrameGroupBy</b> object.

To narrow our focus, we are only going to keep the fullname, gender, and inches_only columns

```python
celebrity_height_df["inches_only"] = celebrity_height_df["feet"] * 12 + celebrity_height_df["inches"]

grouped_by_first_name = celebrity_height_df.groupby("firstname")[["firstname", "fullname", "gender", "inches_only"]]
```

Let's now examine our new grouping.

```python
print("Grouped By First Name DataFrame Head:")
print()
print(grouped_by_first_name.head())
print()
print()
print("Grouped By First Name Dataframe Tail:")
print()
print(grouped_by_first_name.tail())
```

It's hard to tell what's going on just from the above. Python has created a group for each unique 'firstname' entry and assigned each row to its appropriate group.

We can, however, perform operations on the groupings. For example, we can find the average height of each celebrity of a given first name using the <b>mean()</b> function.

Note that we use the flag: <b>numeric_only = True</b>. This tells Python, that we only want to average values that are numbers.

If we try to average values that are not numbers, we get nonsense values.

```python
average_height_by_name = celebrity_height_df.groupby("firstname").mean(numeric_only = True)

average_height_by_name
```

We can instead find the shortest celebrity with each name by using the <b>min()</b> function

```python
shortest_by_name = celebrity_height_df.groupby("firstname").mean(numeric_only = True)

shortest_by_name
```

Or, we can find the tallest celebrity with each name by using the <b>max()</b> function.

```python
tallest_by_name = celebrity_height_df.groupby("firstname").mean(numeric_only = True)

tallest_by_name
```

We can even make up our own functions to parse the data by using the <b>agg()</b> function which stands for "aggregate".

For example, lets create a function that gives us the ratio of tallest to shortest celebrity for each first name.

```python
def ratio_of_tallest_to_shortest(height_list):
    tallest = max(height_list)
    shortest = min(height_list)
    return tallest / shortest

height_ratio_by_name = celebrity_height_df.groupby("firstname")[["inches_only"]].agg(ratio_of_tallest_to_shortest)

height_ratio_by_name = height_ratio_by_name.rename(columns = {"inches_only" : "height_ratio"})

height_ratio_by_name
```

Now that we have this new information, we can ask other questions, such as: "What is the biggest ratio for any name?" and "Which name has the biggest ratio?"

```python
# What is the biggest ratio?

max_ratio = height_ratio_by_name.max()[0]

print("Max Ratio is: " + str(max_ratio))
print()

# What name has this ratio?

height_ratio_by_name.query("height_ratio == @max_ratio")
```

Another thing that we can do is rearrange the information in our DataFrame based on a condition that we set using the <b>sort_values()</b> function.

To show this, lets sort our celebrities by their height as listed in the "inches_only" column.

```python
celebrity_height_df.sort_values("inches_only")
```

In the above example, we sorted the values from least to greatest, we can do the opposite by setting the flag: <b>ascending = False</b>.

```python
celebrity_height_df.sort_values("inches_only", ascending = False)
```

Now, we have a new way of finding the shortest or tallest celebrity with a given name by sorting the DataFrame and finding the first time a particular name appears.

For example, lets find the shortest celebrity with each name by using <b>agg()</b> to find the first time a name appears in a list sorted from shortest to tallest.

```python
def get_first_item(list_of_names):
    return list_of_names.iloc[0]

celebrity_height_df.sort_values("inches_only").groupby("firstname").agg(get_first_item)
```

As with anything in programming, there is more than one way to do a task.

Let us repeat the above using the <b>groupby()</b> and <b>first()</b> functions.

```python
celebrity_height_df.sort_values("inches_only").groupby("firstname").first()
```

We can even do some silly things like add up all the heights of each celebrity for a given first name using <b>sum()</b>.

```python
# What is the total height of each celebrity name?

total_height = celebrity_height_df.groupby("firstname").sum(numeric_only = True)

total_height
```

We can also do something more sensible, like ask how many celebrities are there in our data of a given name using <b>value_counts()</b>.

```python
name_count = celebrity_height_df["firstname"].value_counts()

name_count
```

Now, let's suppose that we want to do something more complicated.

Say that we want to compute the ratio of each celebrity's height to the average height of other celebrities with their same first name.

```python
average_height = celebrity_height_df.groupby("firstname").mean(numeric_only = True)

celebrities_indexed_by_name = celebrity_height_df.set_index("firstname")

fraction_height_df = celebrities_indexed_by_name / average_height

fraction_height_df
```

---

In the above example, we divided all the values in "celebrities_indexed_by_name" with all the values in "average_height".

Whenever there was a numeric value, the result was straight forward. However, non-numeric values can't be divided and thus gave us "NaN" or "Not a Number".

Let's redo that, but instead of dividing all of the values, lets just divide the "inches_only" columns and create a new column with the ratio.

To accomplish this, we are first going to use <b>reindex()</b> to organize the DataFrame based on firstname and fullname.

---

```python
df_indexed_by_firstname_and_fullname = celebrity_height_df.set_index(["firstname", "fullname"])
df_indexed_by_firstname_and_fullname
```
