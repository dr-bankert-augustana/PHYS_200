<font size=5><b>3.1 Creating DataFrames From Dictionaries</b></font>

<p align="justify">
One of the most straightforward ways to create a DataFrame in Pandas is by using a Python dictionary. In this approach, each key in the dictionary represents a column name, and the corresponding value is a list (or array) containing the data for that column. Pandas automatically aligns these values by row, as long as the lists are of equal length.
</p>

<br>

<p align="justify">
This method is especially useful when you already have structured data in Python (for example, experimental measurements or simulation results stored in variables) and you want to organize them into a tabular format for further analysis.
</p>

<br>

<b>Code Example:</b>

In the following code cell, we create a simple DataFrame from some synthetic experimental data.

```python
# Import the Pandas Library:

import pandas as pd

# Create a dictionary of data:

data = {
    "Time (s)": [0, 1, 2, 3, 4],
    "Position (m)": [0, 2.5, 10, 22.5, 40],
    "Velocity (m/s)": [0, 5, 10, 15, 20]
}

# Convert the dictionary into a DataFrame:

df = pd.DataFrame(data)

# Display the DataFrame:

display(df)
```

<b>Understanding the Output</b>

<p align="justify">
This printed table looks very similar to what you'd see in a spreadsheet like Excel or Google Sheets:
</p>

  * <b>Columns:</b> Each column corresponds to one of the keys from the dictionary. In this case, we have three columns: <code>Time (s)</code>, <code>Position (m)</code>, and <code>Velocity (m/s)</code>. These serve the same purpose as column headers in a spreadsheet, labeling the type of data in each column.

  * <b>Rows:</b> Each row contains a single set of related data points; one time value, one position, and one velocity. These rows represent observations or measurements taken at different points in time.

  * <b>Index column (on the far left):</b> By default, Pandas automatically adds an index column starting from 0. This index uniquely identifies each row, similar to row numbers in a spreadsheet. While it doesn't carry physical meaning in this example, it becomes very useful when accessing or referencing specific rows programmatically.

<br>

<p align="justify">
So, in essence, the Pandas DataFrame behaves much like a digital spreadsheet, but with the added power of Python. As we see in this lesson, you can apply mathematical operations, filter rows, compute statistics, and generate plots all from this structured format, making it a key tool in your computational physics toolkit.
</p>

<font size=5><b>3.2 Creating a DataFrame from a CSV File</b></font>

<p align="justify">
In real-world physics work, it's common to store data in external files; especially CSV (Comma-Separated Values) files. These are simple text files where each line represents a row of data, and values are separated by commas. Think of it as a stripped-down spreadsheet that can be easily shared and read by many programs. In fact, you can export a spreadsheet from any common program (such as Excel, Google Sheets, etc) as a CSV file.
</p>

<br>

<p align="justify">
Pandas makes it incredibly easy to read data from a CSV file and convert it into a DataFrame using the <code>read_csv()</code> function.
</p>

<br>

<b>Code Example:</b>

In the following code cell, we read in the data from a CSV file into a Pandas DataFrame.

```python
# Set the location of the CSV file (can be local or a website):

url = "https://raw.githubusercontent.com/dr-bankert-augustana/PHYS_200/refs/heads/main/Data/Module_1/celebrity-heights.csv"

# Turn the CSV data into a Pandas DataFrame:

celebrity_height_df = pd.read_csv(url)

# Display the DataFrame:

display(celebrity_height_df)
```

<font size = 6> <b> 4. Exploring a DataFrame </b> </font>

---

<font size=5><b>4.1 Methods for Exploring the DataFrame</b></font>

<p align="justify">
Once you've loaded data into a Pandas DataFrame (whether from a dictionary, CSV file, or another source) it's important to understand what's in it and whether it looks right before jumping into analysis. Pandas provides several built-in methods to help us quickly explore the dataset.
</p>

<center>$\rule$</center>
<font size=5><b>4.2 <code>info()</code>: Structure and Metadata</b></font>

This method gives you a concise summary of the DataFrame’s structure, including:

  * The number of rows and columns

  * Column names and data types (int64, float64, object, etc.)

  * Number of non-null (i.e. non-missing) values in each column

  * Memory usage

<br>

<p align="justify">
<b>Why it's useful:</b> Helps you quickly check for missing data and see if the data types are appropriate (e.g. no text where numbers should be).
</p>

<br>

<b>Code Example:</b>

In the following code cell, we use <code>info()</code> to look at our celebrity heights DataFrame.

```python
# Display the Output of info():

display(celebrity_height_df.info())
```

<center>$\rule$</center>
<font size=5><b>4.3 <code>df.describe()</code>: Summary Statistics</b></font>

This method computes common statistical summaries for each numeric column, such as:

* Count

* Mean

* Standard deviation

* Min and max

* 25th, 50th (median), and 75th percentiles

<br>

<b>Why it's useful</b>: Provides a quick overview of the data's distribution and helps spot unusual values (e.g., extreme outliers or zero variance).

<br>

<b>Code Example:</b>

In the following code cell, we use <code>describe()</code> to look at our celebrity heights DataFrame.

<center>$\rule$</center>
<font size=5><b>4.4 <code>head(n)</code>: Display First n Rows (Default: 5)</b></font>

Displays the first few rows of the DataFrame. This is often your first stop after loading data to verify that everything looks as expected (column names, data format, etc.).

<br>

<b>Code Example:</b>

In the following code cell, we use <code>head(n)</code> to look at our celebrity heights DataFrame.

```python
# Display the first 5 rows of the celebrity heights DataFrame:

print("First 5 rows:")

display(celebrity_height_df.head())

print()

# Display the first 10 rows of the celebrity heights DataFrame:

print("First 10 rows:")

display(celebrity_height_df.head(10))
```

<center>$\rule$</center>
<font size=5><b>4.5 <code>tail(n)</code>: Display Last n Rows (Default: 5)</b></font>

Displays the last few rows of the DataFrame. This is useful to:

  * See the end of a time series

  * Verify that data loaded completely

  * Check for anomalies or formatting issues at the bottom of the file

<br>

<b>Code Example:</b>

In the following code cell, we use <code>tail(n)</code> to look at our celebrity heights DataFrame.

```python
# Display the last 5 rows of the celebrity heights DataFrame:

print("Last 5 rows:")

display(celebrity_height_df.tail())

print()

# Display the last 10 rows of the celebrity heights DataFrame:

print("Last 10 rows:")

display(celebrity_height_df.tail(10))
```

<center>$\rule$</center>
<font size=5><b>4.6 Other Useful Commands For Investigating DataFrames</b></font>

  * <b>df.columns</b>:
  
    * Returns a list-like object containing the column labels.

    * Useful for checking column names or looping through columns.

  * <b>df.index</b>:

    * Returns the row index of the DataFrame (default is a RangeIndex).

    * Important if you're dealing with custom indices like timestamps or labels.

  * <b>df.shape</b>:

    * Returns a tuple representing the dimensions of the DataFrame: (rows, columns).

  * <b>df.dtypes</b>:

    * Shows the data type of each column (e.g., float64, int64, object).

    * Helpful when debugging unexpected behavior in calculations.

  * <b>df.nunique()</b>:

    * Returns the number of unique values in each column.

    * Good for identifying categorical vs. continuous data, or detecting duplicates.

  * <b>df.isnull().sum()</b>:

    * Identifies missing values by column.

    * Essential for cleaning data before analysis or plotting.

  * <b>df.sample(n)</b>:

    * Returns a random sample of n rows from the DataFrame.

    * Great for large datasets when you just want to inspect a few random entries.

  * <b>df["column_name"].value_counts()</b>:

    * For a single column, shows how often each unique value appears (useful for categorical or discrete data).

<font size = 6> <b> 5. Selecting Data in a DataFrame </b> </font>

---

<font size=5><b>5.1 Bracket Notation []</b></font>

<p align="justify">
Once we have loaded and explored our DataFrame, the next step is accessing specific columns or subsets of data. One of the most common and flexible ways to do this is using bracket selection, which looks like: <code>DataFrame[selector]</code>
</p>

<br>

<p align="justify">
Depending on what we put inside the brackets, we can select a column, multiple columns, or filter rows using logical conditions.
</p>

<center>$\rule$</center>
<font size=5><b>5.2 Selecting a Single Column</b></font>

<p align="justify">
If we pass a string (column name) inside the brackets, Pandas returns a Series (a one-dimensional view of that column). This works like selecting a column in a spreadsheet by its header. The result behaves like a list but with labels (index) attached.
</p>

<br>

><b>Note:</b><p align="justify">
We can also call up a column by using: <code>DF_Name.Column_Name</code>
</p>

<br>

<b>Code Example:</b>

```python
# Select the 'firstname' column of the celebrity heights data:

first_names = celebrity_height_df['firstname']

# Display the firstnames data:

display(first_names)

# Select the 'lastname' column of the celebrity heights data:

last_names = celebrity_height_df.lastname

# Display the lastnames data:

display(last_names)
```

<center>$\rule$</center>
<font size=5><b>5.3 Selecting a Multiple Columns</b></font>

<p align="justify">
To select more than one column, pass in a list of column names inside the brackets: <code>subset_df = df[[column_1, column_2]]</code>. Now you get a new DataFrame with just those columns. Similar to hiding other columns in a spreadsheet.
</p>

<br>

<b>Code Example:</b>

```python
# Select the 'firstname' and 'lastname' columns of the celebrity heights data:

names = celebrity_height_df[['firstname', 'lastname']]

# Display the subset data:

display(names)
```

<center>$\rule$</center>
<font size=5><b>5.4 Selecting Rows with Logical Conditions</b></font>

<p align="justify">
You can also use bracket selection with boolean expressions to filter rows. This is incredibly powerful for working with physical measurements or simulation outputs. This is called using a <b>Boolean Mask</b>. Example: <code>subset = df[boolean_mask]</code>. This filters the DataFrame and only returns rows that meet the condition, just like using filters in Excel or Google Sheets.
</p>

><b>Note:</b><p align="justify">
You can build more complex filters by combining multiple conditions.
</p>

<br>

<b>Code Example:</b>

```python
# Create a subset of people who are over 2.2 meters tall:

tall_people = celebrity_height_df[celebrity_height_df['meters'] > 2.2]

# Display the subset data:

print("Tall People:")
display(tall_people)
print()

# Create a subset of people who are between 1.6 and 1.7 meters tall:

medium_tall_people = celebrity_height_df[(celebrity_height_df['meters'] > 1.6) & (celebrity_height_df['meters'] < 1.7)]

# Display the subset data:

print("Medium Tall People:")
display(medium_tall_people)
```
