<font size = 5> <b> 2.3 What is Computer Coding? </b> </font>

Computer code is a detailed, explicit set of instructions that tells a computer what to do. A computer program is simply a list of these instructions, or statements, which are executed in order; typically from top to bottom, unless modified by control flow statements (e.g., loops, conditionals; topics that we will cover soon) that change the order or repeat certain parts.

<br>

Computer coding is the process of writing these instructions. The exact way you write them depends on the programming language you're using. Once written, your code is passed through a compiler or interpreter, which translates the human-readable instructions into machine-readable code; binary instructions that the computer's processor can understand and execute. No matter which language you write in, the end goal is the same: to give the computer a precise set of instructions in a form it can execute.

<center>$\rule$</center>

<font size = 5> <b> 2.4 Why Use Python? </b> </font>

If all programming languages eventually produce the same kind of machine code, why choose one over another? The short answer: <b>ease of use and project needs</b>.

<br>

While many languages can express the same concepts, they do so with different levels of complexity and control. Here's the key distinction:

  * <b>Low-level languages</b> like <b>C++</b> or <b>Fortran</b> give programmers fine-grained control over how the computer executes instructions. This makes them ideal for building programs that need to be extremely fast and efficient. However, they require a lot of careful work from the programmer; you must be very specific and precise in every step.

  * <b>High-level languages</b> like <b>Python</b> are designed to be more user-friendly and readable. They handle many low-level details behind the scenes, which makes them:

    * Easier to learn

    * Quicker to write

    * More forgiving when you're just starting out

<br>

This added simplicity comes with a trade-off: Python programs may run a bit more slowly or be less efficient than ones written in lower-level languages. But for teaching, prototyping, and even simple to moderate scientific computing, that trade-off is almost always worth it.

<br>

><b>In this class, we'll use Python because it's beginner-friendly, widely used in science and engineering, and powerful enough to handle everything from simple calculations to complex simulations.</b>
