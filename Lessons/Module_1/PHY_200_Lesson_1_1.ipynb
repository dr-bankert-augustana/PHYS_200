<font size = 5> <b> 2.2 What is Computer Coding? </b> </font>

Computer code is a detailed, explicit set of instructions which tell the computer what to do. A computer program is a list of statements. When you run the program, the statements are carried out in order, usually from top to bottom unless you put in control flow statements to change the order (for example, do some statements several times before proceeding). "Computer Coding" is the act of creating these sets of instructions. The precise manner in which you 'word' your instructions depends on the computer language that you are using. The instructions you write are then read by a [compiler](#Definition-Compiler) which translates the human-readable code that you produced into machine-readable code that the computer's processor can understand. Regardless of what human-readable computer language you use to write in, the instructions will be translated into the same machine-readable format by a compiler.

<center>$\rule$</center>

<font size = 5> <b> 2.2 Why Use Python? </b> </font>

If the end result is the same machine-readable code, why does it matter which human-readable code language we use? The simple answer is that it doesn't, not really (just as you can express the same concepts and ideas in essentially any spoken language). Ultimately, the computer coding language that you choose to use comes down to preference, and utility. Some languages (like C++ or Fortran) provide users with extremely fine control over how the code is executed once its translated into machine-readable form. This allows them to create lighting-fast programs (we call these 'low-level' languages). The trade-off is that these languages require a lot more work on the part of the user to be extremely specific about what they want the computer to do. Other languages (like Python), have a lot more overhead or "behind the scenes" work going on in the compiler. This means that the programs may not be as efficient when translated into computer-readable code, but they are much easier to write and more forgiving to the user who doesn't need to be as pendantic when writing their commands (we call these 'higher-level' languages).
