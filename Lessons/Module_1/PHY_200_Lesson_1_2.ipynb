<font size = 6> <b> 3. Conditional Statements</b> </font>

---

<font size = 5> <b> 3.1 What are Conditional Statements? </b> </font>

In many cases, we want a program to make decisions based on certain conditions. For a real-life example, suppose we have a bucket of different kinds of fruit, but we only want to pick out the oranges.

<br>

To a human, we might say: <b>"Only take out the oranges."</b>

<br>

To a computer, we must be more specific: <b>"Look at each fruit. If the fruit is an orange, then take it out."</b>

<br>

This kind of decision-making is done in programming using <a href = #Definition-Conditional></b>conditional statements</b></a>. These are instructions that tell the computer:

  * check whether or not some given given statement is <b>True</b> what to check, and ...

  * what to do <b>if</b> the check is <b>True</b>
  
  * optionally, what to do <b>if</b> it is <b>False</b>.

<center>$\rule$</center>
<font size = 5> <b> 3.2 If Statements </b> </font>

Creating a conditional is often (though not always) done by using an <a href = #Definition-If></b>if statement</b></a>.

<br>

<b> Code Example: </b>

In the following code cell, we look at an example of a simple conditional statement in Python.

```python
# Create two variables and use if statements to check if their value exceeds 5.0.

value_1 = 3.0
value_2 = 7.0

##=============================================================================================##
## Value_1:                                                                                    ##
##=============================================================================================##

# Check whether (value_1 is greater than 5.0) is a True statement.

if (value_1 > 5.0):

  # If (value_1 is greater than 5.0) is True, the following code is executed.

  print("Value_1 is greater than 5.0.")

##=============================================================================================##
## Value_2:                                                                                    ##
##=============================================================================================##

# Check whether (value_2 is greater than 5.0) is a True statement.

if (value_2 > 5.0):

  # If (value_2 is greater than 5.0) is True, the following code is executed.

  print("Value_2 is greater than 5.0.")
```

So, what's happening in this code?

  * <code> if (value_1 > 5.0)</code> → This is the condition. Our program is checking if this condition is <b>True</b>.

  * <code> print(...)</code> → This is the action. It runs only if the program evaluates condition and finds it to be <b>True</b>.

><b>Note:</b> Indentation matters! In Python, the code block under the if statement must be indented (typically 4 spaces or 1 tab). This "indent-block" tells Python what code is intended to be run <b>if</b> the condition is <b>True</b>. <b>If</b> you don't indent properly, You'll likely get an <code>IndentationError</code>

<br>

<center>$\rule$</center>
<font size = 5> <b> 3.3 If-Else Statements </b> </font>

<p align="justify">
We now know how to set up the logic for a program to run a block of code <b>if</b> a given condition is met. Now, we want to set up the logic to specify what happens both when a conditional is met (<b>True</b>) and when it is not met (<b>False</b>). In Python, this is done by adding an <a href = #Definition-Else></b>else statement</b></a> to our existing if statement.
</p>

<br>

<b> Code Example: </b>

Let's revisit our previous example and define some behavior for the program in the circumstance that a number is not not greater than 5.0.

```python
# Create two variables and use if-else statements to define the program's behavior
# when their value exceeds 5.0 and when it does not.

value_1 = 3.0
value_2 = 7.0

##=============================================================================================##
## Value_1:                                                                                    ##
##=============================================================================================##

# Check whether (value_1 is greater than 5.0) is a True statement.

if (value_1 > 5.0):

  # If (value_1 is greater than 5.0) is True, the following code is executed.

  print("Value_1 is greater than 5.0.")

# Define the behavior if (value_1 greater than 5.0) is False:

else:

  # If (value_1 is greater than 5.0) is False, the following code is executed.

  print("Value_1 is not greater than 5.0.")

##=============================================================================================##
## Value_2:                                                                                    ##
##=============================================================================================##

# Check whether (value_2 is greater than 5.0) is a True statement.

if (value_2 > 5.0):

  # If (value_2 is greater than 5.0) is True, the following code is executed.

  print("Value_2 is greater than 5.0.")

# Define the behavior if (value_2 greater than 5.0) is False:

else:

  # If (value_2 is greater than 5.0) is False, the following code is executed.

  print("Value_2 is not greater than 5.0")
```

><p align="justify"><b>Reminder:</b> Computers don't understand meaning like we do. They need explicit instructions. That's why we have to:
>
> * Clearly define the condition to check,
> * Specify exactly what to do when the condition is met,
> * And define what to do when it is not.
</p>

<br>

<center>$\rule$</center>
<font size = 5> <b> 3.4 Elif Statements </b> </font>

We can also program in behavior for more than one condition. In Python, this is done with the <a href = #Definition-Elif></b>elif statement</b></a>. <b>Elif</b> stands for "else if" and what it means is <b>"if the previous condition is found False, test this new condition"</b>.

<br>

<b> Code Example: </b>

In the following Code cell, we add some elif statements onto our previous example. We are now checking to see if the numbers are equal to 5.0.

><b>Note:</b> In Python, the <a href = #Definition-Boolean></b>boolean</b></a> (A value that is either <b>True</b> or <b>False</b>) operator to check equality is two equal signs: "==".

```python
# Create three variables and use if-elif-else statements to define the program's behavior
# when their value exceeds 5.0 and when it does not.

value_1 = 3.0
value_2 = 7.0
value_3 = 5.0

##=============================================================================================##
## Value_1:                                                                                    ##
##=============================================================================================##

# Check whether (value_1 is greater than 5.0) is a True statement.

if (value_1 > 5.0):

  # If (value_1 is greater than 5.0) is True, the following code is executed.

  print("Value_1 is greater than 5.0.")

# If (value_1 is greater than 5.0) is False, check if (value_1 equals 5.0) is a True Statement.

elif (value_1 == 5.0):

  # If (value_1 equals 5.0) is True, the following code is executed.

  print("Value_1 is equal to 5.0.")

# Define the behavior if (value_1 greater than 5.0) and (value_1 equals 5.0) are both False:

else:

  # If (value_1 is greater than 5.0) is False, and (value_1 equals 5.0) is False,
  # the following code is executed.

  print("Value_1 is not greater than 5.0.")

##=============================================================================================##
## Value_2:                                                                                    ##
##=============================================================================================##

# Check whether (value_2 is greater than 5.0) is a True statement.

if (value_2 > 5.0):

  # If (value_2 is greater than 5.0) is True, the following code is executed.

  print("Value_2 is greater than 5.0.")

# If (value_2 is greater than 5.0) is False, check if (value_2 equals 5.0) is a True Statement.

elif (value_2 == 5.0):

  # If (value_2 equals 5.0) is True, the following code is executed.

  print("Value_2 is equal to 5.0.")

# Define the behavior if (value_2 greater than 5.0) and (value_2 equals 5.0) are both False:

else:

  # If (value_2 is greater than 5.0) is False, and (value_2 equals 5.0) is False,
  # the following code is executed.

  print("Value_2 is not greater than 5.0")

##=============================================================================================##
## Value_3:                                                                                    ##
##=============================================================================================##

# Check whether (value_3 is greater than 5.0) is a True statement.

if (value_3 > 5.0):

  # If (value_3 is greater than 5.0) is True, the following code is executed.

  print("Value_3 is greater than 5.0.")

# If (value_3 is greater than 5.0) is False, check if (value_3 equals 5.0) is a True Statement.

elif (value_3 == 5.0):

  # If (value_3 equals 5.0) is True, the following code is executed.

  print("Value_3 is equal to 5.0.")

# Define the behavior if (value_3 greater than 5.0) and (value_3 equals 5.0) are both False:

else:

  # If (value_3 is greater than 5.0) is False, and (value_3 equals 5.0) is False,
  # the following code is executed.

  print("Value_3 is not greater than 5.0")
```

<font size = 6> <b> 4. Loops</b> </font>

---

<font size = 5> <b> 4.1 What are Loops? </b> </font>

Sometimes, we need to repeat a task many times. For example, you might want to:

  * print every number from 1 to 10

  * calculate the position of a particle at every time step

  * check through every item in a list or array

<br>

Doing this manually would be repetitive and time-consuming. Instead, we use a logic structure called a <a href = #Definition-Loop></b>loop</b></a>, which tells the computer to repeat a task; either for every item in a group, or until a given condition is met.

<br>

There are two main types of loops in Python:

  * <a href = #Definition-For></b>for loops</b></a>; used when looping over items in a sequence

  * <a href = #Definition-While></b>while loop</b></a>; used when repeating until a given condition becomes False

<center>$\rule$</center>
<font size = 5> <b> 4.2 For Loops </b> </font>

Let's start with for loops. A for loop is used to repeat a block of code for each item in a given list (or other sequence). This is useful when you know precisely how many repetitions you want to loop through.

<b>Code Example:</b>

Let's see how a for loop can help us iterate through a list.

```python
# Create a list containing the names of different types of particles.

particles = ["proton", "neutron", "electron"]

# Loop through each of the items in the list and print them.

for particle in particles:

  # This code will run once for each item in the list.

  print("Particle:", particle)
```

So, what's happening here?

  * <code>for particle in particles:</code> → We're telling the computer to loop through each item in the list.

  * <code> print(...)</code> → This action is repeated once for each item in the list during the loop.

<br>

<center>$\rule$</center>
<font size = 5> <b> 4.3 range() </b> </font>

Often, we want to loop through a specific number of iterations; for example, we want to run a code block 10 times. Python gives us the <code>range()</code> function for just this purpose.

<br>

<b>Code Example:</b>

In the following code cell, we will make use of <code>range()</code> in our for loop.

<br>

><b>Note:</b> The command <code>range(value)</code> generates the numbers: 0, 1, 2, 3, ... value - 1. (<b>Not value</b>). The sequence provided ends just before the value you provide. This is because Python uses zero-based indexing, which is common in programming.

```python
# Use range(5) to produce the sequence 0, 1, 2, 3, 4 and at each iteration, the variable
# i takes on the current value of the sequence. At the end of the iteration, i takes on the
# next value in the sequence.

for i in range(5):

  # Print out the current value of the variable i:

  print("i =", i)
```

<b>Code Example:</b>

In the previous code cell, we saw the function range(5) generates the numbers: 0, 1, 2, 3, 4 (<b>not 5</b>). The loop runs 5 times, and on each pass, the variable i takes on the next value in the sequence. If we want to start at a number other than 0, we use: range(start, stop). If we want to iterate by an amount other than 1, we use: range(start, stop, step). In the next code cell, we see how we can change the start, stop, and step of <code>range()</code>.

```python
##=============================================================================================##
## Loop 1:                                                                                     ##
##=============================================================================================##

# Print the time at each second for a 10-second simulation, starting at 0 seconds.

print("Loop 1:")
print()

for t in range(11):  # From 0 to 10 seconds

  print("Time:", t, "seconds")

##=============================================================================================##
## Loop 2:                                                                                     ##
##=============================================================================================##

# Print the time at each second for a 12-second simulation, starting at 3 seconds.

print()
print("Loop 2:")
print()

for t in range(3, 16):  # From 3 to 15 seconds

  print("Time:", t, "seconds")

##=============================================================================================##
## Loop 3:                                                                                     ##
##=============================================================================================##

# Print the time at every other second for a 10-second simulation, starting at 5 seconds.

print()
print("Loop 3:")
print()

for t in range(5, 16, 2):  # Odd values from 5 to 15 seconds

  print("Time:", t, "seconds")


##=============================================================================================##
## Loop 4:                                                                                     ##
##=============================================================================================##

# Print the time at every second for a 10-second countdown starting at 10 seconds.

print()
print("Loop 4:")
print()

for t in range(10, -1, -1):  # From 10 to 0 seconds

  print("Time:", t, "seconds")
```

><b>Tip:</b> It is common to use the letters <b>i</b>, <b>j</b>, and <b>k</b> as iteration variables in for loops when using <code>range()</code>.

<br>

<b>Nested Loops:</b>

<p align="justify">
It is possible to have loops within loops. These are called "nested" loops. In a nested loop, the inner loop runs its full course every time the outer loop runs an iteration. This is very useful if we have multi-dimensional data that we need to examine or if we have a smaller task that needs to be performed as part of a larger task.
</p>

<b>Code Example:</b>

In the following code cell, we see an example of a nested set of loops and how the inner loop runs fully each time the outer loop iterates.

```python
# Here, we have an example of nested for loops. The outer loop iterates the variable i,
# and the inner loop iterates the variable j.

# The outer loop runs i = 0, 1, 2:

for i in range (0, 3):

  # The inner loop runs j = 0, 1, 2:

  for j in range(0, 3):

    # Print the value of i for each iteration of the outer loop:
    # Print the value of j for each iteraiton of the inner loop:

    print("Outer Loop:", i, "; Inner Loop:", j)

    # At the end of the inner loop, print a newline:

    if (j == 2):

      print()
```

<center>$\rule$</center>
<font size = 5> <b> 4.4 While Loops </b> </font>

A while loop repeats as long as a provided condition is True. This is useful when we don't know in advance how many iterations we need to loop through.

<br>

><b>WARNING:</b><p align="justify"> While loops are more versatile than for loops, but they are also more dangerous. It is very easy to accidentally create an infinite loop using while <b>by providing a condition that is always True</b>. This is bad because it can cause your program to freeze or crash, especially if it runs indefinitely without a way to stop.</p>

<br>

<b>Code Example:</b>

In the following code cell, we construct a while loop that executes a block of code <b>while</b> a particular value is less than 5. During each iteration of the loop, the value is increased. Once the value becomes greater than 5, the code block will no longer execute.

```python
# Start the value of the variable at 0.

counter = 0

# Run the loop *while* the statement (counter < 5) is True:

while (counter < 5):

  # Display the current value of the counter.

  print("The value of the counter is", counter)

  # Increase the value of the counter by 1:

  counter += 1
```

So, what's happening?

  * <code>(counter < 5)</code> → this condition is checked before each loop.

  * The loop continues until the condition becomes False.

  * <code>counter += 1</code> → increases the value of counter by 1 each iteration.

  * <b>If you forget to update the condition, the loop could run forever!</b>


><b>Note:</b><p align="justify"> The value of counter is never printed at 5. This is because the provided condition is checked <b>before</b> executing the code block. In the final iteration of the code block, the value of counter becomes 5, thus when the statement (counter < 5) is evaluated at the beginning of the next iteration, it is found to be False and the loop exits without executing the current iteration.</p>

<br>

<b>Code Example:</b>

In the next code cell, we add a second print statement to show how the value of counter is updated and how that affects the execution of the code block.

```python
# Start the value of the variable at 0.

counter = 0

# Run the loop *while* the statement (counter < 5) is True:

while (counter < 5):

  # Display the old value of the counter.

  print("The old value of the counter is", counter)

  # Increase the value of the counter by 1:

  counter += 1

  # Display the new value of the counter.

  print("The new value of the counter is", counter)
  print()
```

<center>$\rule$</center>
<font size = 5> <b> 4.5 What Kind of Loop Should I Use? </b> </font>

Use a <b>for loop</b> when:

  * You're looping over items in a list

  * You're using a known number of iterations

<br>

Use a <b>while loop</b> when:

  * You want to repeat until a condition becomes False

  * You don't know in advance how many times to loop

---

<font size = 5> <b> 5.1 What Are Functions? </b> </font>

<p align="justify">
Computer coding often relies upon repetition. Rather than rewriting the same block of code over and over, we can group that code into a <a href = #Definition-Function></b>function</b></a>; a named block of reusable code.
Functions are very, very useful tools in computer programming. Using functions makes your code:</p>

  * Easier to read

  * Easier to reuse

  * Easier to debug and maintain

<br>

You've actually already been using functions! Examples include:

  * print() — prints information to the screen

  * range() — creates a sequence of numbers

<br>

<b>Code Example:</b>

In Python, you create a function using the keyword <b>def</b>, followed by the function name and a set of parentheses. Now, let's practice writing our own functions.

```python
# Define a function called greet that prints a welcome message.

def greet():

  # Here is the code that executes when the function is called.

  print("Hello, and welcome to computational physics!")

# Call (run) the function:
greet()
```

What's Happening Here?

  * <code>def greet()</code> → This defines a function named greet.

  * The indented block below it is the code that will run when the function is called.

  * <code>greet()</code> → This line actually runs (or calls) the function.

<br>

><b>Note:</b> It is best practice to give functions short, but descriptive names. This makes reading a program's code much easier for you and for others.

<center>$\rule$</center>
<font size = 5> <b> 5.2 Functions with Parameters </b> </font>

Functions become much more powerful when we add input parameters; inputs that let us pass information into the function.

<br>

<b>Code Example:</b>

In the following code cell, we create a function that takes an input which is passed to the code block housed within the function's definition.

```python
# Define a function that takes in a name and prints a greeting.

def personalized_greet(name):

    print("Hello,", name + "! Welcome to computational physics.")

# Call the function with different inputs:

personalized_greet("Alex")
print()
personalized_greet("Jordan")
```

What's Happening Here?

  * <code>name</code> is a parameter; a variable that the function uses when it's called.

  * When we call personalized_greet("Alex"), it prints a message using that name.

<br>

><p align="justify"> <b>Note:</b> The variable name within the function does not need to be the same name as the variable fed to the function by the program. The name that you use when you define the function is just for internal use within the function. (This is called variable scoping) </p>

<center>$\rule$</center>
<font size = 5> <b> 5.3 Functions That Return A Value </b> </font>

So far, our functions have just printed results. But sometimes we want a function to calculate something and give us the result back. For this, we use the return keyword.

```python
# Define a function that calculates the square of a number and returns the result.

def square(x):

  result = x ** 2

  return result

# Store the returned value in a variable:
squared_value = square(4)

# Print the result:
print("The square of 4 is:", squared_value)
```

What's Happening Here?

  * <code>square(x)</code> takes an input x and calculates its square.

  * <code>return result</code> sends that value back to wherever the function was called.

  * <code>squared_value = square(4)</code> stores the returned result in a variable.

<center>$\rule$</center>
<font size = 5> <b> 5.4 Functions With Multiple Input Parameters </b> </font>

Functions can take more than one parameter. This lets us write general-purpose functions that work with a wide range of inputs.

```python
# Define a function that calculates the final position of an object
# under constant velocity motion: final_position = initial_position + velocity * time

def final_position(initial_position, velocity, time):

  return initial_position + velocity * time

# Call the function with sample values:

position = final_position(0, 3.0, 4.0)

print("The object's final position is:", position, "meters")
```

What's Happening Here?

  * The function takes three parameters: <code>initial_position</code>, <code>velocity</code>, and <code>time</code>.

  * It calculates the final position and returns it.

  * When we call the function with values, it does the math and returns the answer.

<center>$\rule$</center>
<font size = 5> <b> 5.5 When To Use Funcitons</b> </font>

You should use functions when:

  * You have code that gets repeated

  * You want to break up a large program into smaller, understandable parts

  * You want to reuse logic with different inputs

Functions are one of the most important tools in programming, and they will be essential for solving physics problems in this course.

<font size = 6> <b> 6. Classes</b> </font>

---

<p style="text-align: justify">
Imagine that you are trying to write a program for a video game. In this game, the player character has certain attributes (health, skills, etc.). We can code out each of these attributes into our program using variables as we have been doing thus far.
</p>

```python
name = "Steve"
health = 50
attack = 10
dodge  = 5
smarts = 8
```

This works if we have only one character, but what if we have more? Lets say that we have two characters, we could write each of their attributes out by making more variables.

```python
name_1 = "Steve"
health_1 = 50
attack_1 = 10
dodge_1  = 5
smarts_1 = 8

name_2 = "Susie"
health_2 = 50
attack_2 = 8
dodge_2 = 12
smarts_2 = 9
```

This is a bit cumbersome, but workable. Now imagine that we wanted to have dozens of characters. This method quickly becomes non-feasible. What we want to do is group together these attributes into one object that we can then interact with when we want to access them. This grouping of variables (and functions) is called a <a href = #Definition-Class></b>class</b></a>.

```python
# Define the class object:

class character:

    # Define the function used to create an instance of the class:

    def __init__(self, name, health, attack, dodge, smarts):

        # Store the values in internal variables.

        self.name   = name
        self.health = health
        self.attack = attack
        self.dodge  = dodge
        self.smarts = smarts

# Create instances of the class:

player_1 = character("Steve", 50, 10, 5, 8)

player_2 = character("Susie", 50, 8, 12, 9)

# Now we can access the elements of the class by calling them:

print("Player 1's name is: ", player_1.name)
print("Player 2's dodge is: ", player_2.dodge)
```

We can even edit the values of the members of the class by directly calling
    them.

```python
# Define the class object:

class character:

    # Define the function used to create an instance of the class:

    def __init__(self, name, health, attack, dodge, smarts):

        # Store the values in internal variables.

        self.name   = name
        self.health = health
        self.attack = attack
        self.dodge  = dodge
        self.smarts = smarts

    # Define a function to update the character's health when attacked.

    def take_damage(self, damage):

        self.health -= damage

# Create instances of the class:

player_1 = character("Steve", 50, 10, 5, 8)

print("Player 1's health old value is: ", player_1.health)

player_1.take_damage(15)

print("Player 1's health new value is: ", player_1.health)
```

<a name="Definition-Boolean"></a>
* <font size = 4> <b> Boolean: </b> </font>
A data type that can be either True or False. Conditional logic in programming depends on Boolean values.

<a name="Definition-Class"></a>
* <font size = 4> <b> Class: </b> </font>
A template for creating objects that bundle together related data and functions. Classes are used in object-oriented programming to model real-world things like particles or systems.

<a name="Definition-Conditional"></a>
* <font size = 4> <b> Conditional Statement: </b> </font>
A programming instruction that tells the computer to perform an action only if a certain condition is true. Used for making decisions in code.

<a name="Definition-Elif"></a>
* <font size = 4> <b> Elif Statement: </b> </font>
Short for “else if”. Used when you want to check additional conditions if the previous ones were false. You can use multiple elif statements to create a chain of conditions.

<a name="Definition-Else"></a>
* <font size = 4> <b> Else Statement: </b> </font>
Used with an if statement to define what happens when the condition is false. The code in the else-block runs only if the if condition is not met.

<a name="Definition-For"></a>
* <font size = 4> <b> For Loop: </b> </font>
A loop that repeats a block of code for each item in a list or over a specified range. Often used when the number of repetitions is known in advance.

<a name="Definition-Function"></a>
* <font size = 4> <b> Function: </b> </font>
A reusable block of code that performs a specific task. Functions take inputs (parameters) and can return outputs. They help break down complex problems into smaller parts.

<a name="Definition-If"></a>
* <font size = 4> <b> If Statement: </b> </font>
A type of conditional that checks whether a condition is true. If it is, the code inside the if-block runs.

<a name="Definition-Loop"></a>
* <font size = 4> <b> Loop: </b> </font>
A structure that tells the computer to repeat a block of code multiple times. Useful when performing repetitive tasks like simulating multiple time steps.

<a name="Definition-While"></a>
* <font size = 4> <b> While Loop: </b> </font>
A loop that repeats as long as a specified condition is true. Useful when you don't know ahead of time how many repetitions you’ll need.
