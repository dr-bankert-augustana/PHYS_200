---

<font size = 5> <b> 3.1 What are Conditional Statements? </b> </font>

In many cases, we want a program to make decisions based on certain conditions. For a real-life example, suppose we have a bucket of different kinds of fruit, but we only want to pick out the oranges.

<br>

To a human, we might say: <b>"Only take out the oranges."</b>

<br>

To a computer, we must be more specific: <b>"Look at each fruit. If the fruit is an orange, then take it out."</b>

<br>

This kind of decision-making is done in programming using <a href = #Definition-Python></b>conditional statements</b></a>. These are instructions that tell the computer:

  * check whether or not some given given statement is <b>True</b> what to check, and

  * what to do <b>if</b> the check is <b>True</b> (often followed by further instructions on what to do <b>if</b> it is <b>False</b>).

<center>$\rule$</center>
<font size = 5> <b> 3.2 If Statements </b> </font>

Creating a conditional is often (though not always) done by using an <a href = #Definition-Python></b>if statement</b></a>.

<br>

<b> Code Example: </b>

In the following code cell, we look at an example of a simple conditional statement in Python.

```python
# Create two variables and use if statements to check if their value exceeds 5.0.

value_1 = 3.0
value_2 = 7.0

# Check if value_1 is greater than 5.0.

if (value_1 > 5.0):

  # If value_1 > 5.0 is True, the following code is executed.

  print("Value_1 is greater than 5.0.")

# Check if value_2 is greater than 5.0.

if (value_2 > 5.0):

  # If value_2 > 5.0 is True, the following code is executed.

  print("Value_2 is greater than 5.0.")
```

So, what's happening in this code?

  * <code> if (value_1 > 5.0)</code> → This is the condition. Our program is checking if this condition is <b>True</b>.

  * <code> print(...)</code> → This is the action. It runs only if the program evaluates condition and finds it to be <b>True</b>.

><b>Note:</b> Indentation matters! In Python, the code block under the if statement must be indented (typically 4 spaces or 1 tab). This "indent-block" tells Python what code is intended to be run <b>if</b> the condition is <b>True</b>.

<br>

<center>$\rule$</center>
<font size = 5> <b> 3.3 If-Else Statements </b> </font>

We now know how to set up the logic for a program to run a block of code <b>if</b> a given condition is met. Now, we want to set up the logic to specify what happens both when a conditional is met (<b>True</b>) and when it is not met (<b>False</b>). In Python, this is done by adding an <a href = #Definition-Python></b>else statement</b></a> to our existing if statement.

<br>

<b> Code Example: </b>

Let's revisit our previous example and define some behavior for the program in the circumstance that a number is not not greater than 5.0.

```python
# Create two variables and use if-else statements to define the program's behavior
# when their value exceeds 5.0 and when it does not.

value_1 = 3.0
value_2 = 7.0

# Check if value_1 is greater than 5.0.

if (value_1 > 5.0):

  # If value_1 > 5.0 is True, the following code is executed.

  print("Value_1 is greater than 5.0.")

else:

  # If value_1 > 5.0 is False, the following code is executed.

  print("Value_1 is not greater than 5.0.")

# Check if value_2 is greater than 5.0.

if (value_2 > 5.0):

  # If value_2 > 5.0 is True, the following code is executed.

  print("Value_2 is greater than 5.0.")

else:

  # If value_2 > 5.0 is False, the following code is executed.

  print("Value_2 is not greater than 5.0")
```
