<font size = 6> <b> 1. Learning Outcomes </b> </font>

---

By the end of this lesson, students will be able to:

  1. <b>Describe</b> what Python is and <b>explain</b> its role in scientific computing.

  2. <b>Use</b> Google Colab to write, run, and <b>modify</b> Python code in an interactive environment.

  3. <b>Create<b> and </b>assign</b> variables to store numerical and textual data in Python.

  4. <b>Identify</b> and <b>interpret</b> common Python errors and practice basic debugging techniques.

  5. <b>Construct</b> and <b>manipulate</b> one-dimensional arrays (lists) to store and process collections of data.

  6. <b>Access</b> and <b>modify</b> elements of arrays using indexing and slicing.

  7. <b>Explain</b> the structure of two-dimensional arrays and <b>perform</b> basic 2D indexing and slicing operations.

  8. <b>Define</b> dictionaries and <b>explain</b> the difference between key-based and index-based data access.

  9. <b>Create</b>, <b>access</b>, and <b>modify</b> dictionaries to organize labeled data in a flexible way.

  10. <b>Differentiate</b> between lists and dictionaries, and <b>choose</b> the appropriate structure for a given problem.
