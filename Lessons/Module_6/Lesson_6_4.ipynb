### <b> Define Display and Plot Functions </b>

```python
#@title This cell defines the functions: display_dataframes and plot_data

##=============================================================================================##
## Function:  display_dataframes                                                               ##
##                                                                                             ##
## Purpose:   Display multiple DataFrames side-by-side with titles                             ##
##                                                                                             ##
## Input(s):  dataframe_list - List of DataFrames to be displayed                              ##
##            title_list     - List of titles for the displayed DataFrames                     ##
##            n_items        - Number of items to display (optional, default = 5)              ##
##                                                                                             ##
## Output(s): None                                                                             ##
##=============================================================================================##

def display_dataframes(dataframe_list, title_list, n_items = 5, tail = False):

  ##===========================================================================================##
  ## Create a String for Housing the Commands to Be Sent to the display_html() Function:       ##
  ##===========================================================================================##

  html_str = ""

  ##===========================================================================================##
  ## Loop Over the Elements in the item_list and title_list:                                   ##
  ##===========================================================================================##

  for df, title in zip(dataframe_list, title_list):

    # Convert the current dataframe info to html:

    if (tail == True):

      html_df = pd.DataFrame(df).tail(n_items).to_html()

    else:

      html_df = pd.DataFrame(df).head(n_items).to_html()

    ##=========================================================================================##
    ## Wrap title and DataFrames in a Styled HTML <div>:                                       ##
    ##=========================================================================================##

    html_str += "<div style='display: inline-block; margin-right: 20px; vertical-align: top;'>"

    html_str += "<h3 style='text-align: center;'>" + str(title) + "</h3><hr>" + str(html_df)

    html_str += "</div>"

  ##===========================================================================================##
  ## Send the HTML String to the display_html() Function:                                      ##
  ##===========================================================================================##

  display_html(html_str, raw = True)

##=============================================================================================##
## Create a Function to Make Plots:                                                            ##
##=============================================================================================##

def plot_data(data_list, x_list = [], y_list = [], color_list = [], title_list = [],
              x_label_list = [], y_label_list = [], aspect_list = [], legend_list = [],
              overplot = False):

  ##===========================================================================================##
  ## If overplot is True, create an overplot of the graphs:                                    ##
  ##===========================================================================================##

  if (overplot == True):

    # Create the graph:

    x = x_list[0]
    y = y_list[0]

    x_size = 10
    y_size = 10 * aspect_list[0]

    ax = plt.figure(figsize = (x_size, y_size)).add_subplot(111)

    # Set the graph title:

    ax.set_title(title_list[0], fontsize = 14)

    # Set the Graph x-axis label:

    ax.set_xlabel(x_label_list[0], fontsize = 12)

    # Set the Graph y-axis label:

    ax.set_ylabel(y_label_list[0], fontsize = 12)

    # Add grid and the legend:

    ax.grid(True, linestyle = '--', alpha = 0.5)

    # Overplot the remaining data:

    for data, color, label in zip(data_list, color_list, legend_list):

      ax.plot(data[x], data[y], color = color, label = label)

    # Display the plots:

    plt.legend()

    plt.show()

  ##===========================================================================================##
  ## If overplot is False, create a series of graphs:                                          ##
  ##===========================================================================================##

  else:

    # Get the settings for each graph to be made:

    for X, Y, color, title, x_label, y_label, aspect in \
      zip(x_list, y_list, color_list, title_list, x_label_list, y_label_list, aspect_list):

      # Create the graph:

      x_size = 10
      y_size = 10 * aspect

      ax = data_list[0].plot(
        kind    = 'line',
        y       = [Y],
        x       = str(X),
        color   = [color],
        figsize = (x_size, y_size)
      )

      # Set the graph title:

      ax.set_title(title, fontsize = 14)

      # Set the Graph x-axis label:

      ax.set_xlabel(x_label, fontsize = 12)

      # Set the Graph y-axis label:

      ax.set_ylabel(y_label, fontsize = 12)

      # Add grid and suppress the legend:

      ax.grid(True, linestyle = '--', alpha = 0.5)

      legend = ax.legend(title = "Model")

      legend.remove()

    # Display the plots:

    plt.show()
```

### <b> Define Differential Equation Solver Algorithms </b>

```python
#@title This cell defines the Euler, Euler-Cromer, Euler-Richardson, and Runge-Kutta 4th Order DEQ Solver Algorithms.

##=============================================================================================##
## Define the 1D Euler Algorithm Change Function:                                              ##
##=============================================================================================##

def cf_euler_1d(state, parameters, acceleration):

  ##===========================================================================================##
  ## Prepare the Algorithm Values:                                                             ##
  ##===========================================================================================##

  # Create a local copy of the state object:

  state_local = state.copy()

  # Get the time step size:

  dt = parameters['dt']

  # Get the current position:

  x = state_local['Position'].copy()

  # Get the current velocity:

  v = state_local['Velocity'].copy()

  # Get the current acceleration:

  a = acceleration(state_local, parameters)

  # Update the position:

  x += v * dt

  # Update the velocity:

  v += a * dt

  ##===========================================================================================##
  ## Update the State:                                                                         ##
  ##===========================================================================================##

  # Update the local state position:

  state_local['Position'] = x

  # Update the local state velocity:

  state_local['Velocity'] = v

  # Update the local state time:

  state_local['Time'] += dt

  ##===========================================================================================##
  ## Return the Updated State:                                                                 ##
  ##===========================================================================================##

  return state_local

##=============================================================================================##
## Define the 1D Euler-Cromer Algorithm Function:                                              ##
##=============================================================================================##

def cf_euler_cromer_1d(state, parameters, acceleration):

  ##===========================================================================================##
  ## Prepare the Algorithm Values:                                                             ##
  ##===========================================================================================##

  # Create a local copy of the state object:

  state_local = state.copy()

  # Get the time step size:

  dt = parameters['dt']

  # Get the current position:

  x = state_local['Position'].copy()

  # Get the current velocity:

  v = state_local['Velocity'].copy()

  # Get the current acceleration:

  a = acceleration(state_local, parameters)

  # Update the velocity:

  v += a * dt

  # Update the position:

  x += v * dt

  ##===========================================================================================##
  ## Update the State:                                                                         ##
  ##===========================================================================================##

  # Update the local state position:

  state_local['Position'] = x

  # Update the local state velocity:

  state_local['Velocity'] = v

  # Update the local state time:

  state_local['Time'] += dt

  ##===========================================================================================##
  ## Return the Updated State:                                                                 ##
  ##===========================================================================================##

  return state_local

##=============================================================================================##
## Define the 1D Euler Algorithm Function:                                                     ##
##=============================================================================================##

def cf_euler_richardson_1d(state, parameters, acceleration):

  ##===========================================================================================##
  ## Prepare the Algorithm Values:                                                             ##
  ##===========================================================================================##

  # Create a local copy of the state object:

  state_local = state.copy()

  # Get the time step size:

  dt = parameters['dt']

  # Get the current position:

  x = state_local['Position'].copy()

  # Get the current velocity:

  v = state_local['Velocity'].copy()

  # Get the current acceleration:

  a = acceleration(state_local, parameters)

  ##===========================================================================================##
  ## Get the Midpoint Values:                                                                  ##
  ##===========================================================================================##

  # Calculate the midpoint velocity:

  v_mid = v + a * dt / 2

  # Calculate the midpoint position:

  x_mid = x + v * dt / 2

  # Update the local state:

  state_local['Position'] = x_mid

  state_local['Velocity'] = v_mid

  # Calculate the midpoint acceleration:

  a_mid = acceleration(state_local, parameters)

  ##===========================================================================================##
  ## Calculate the Updated Values:                                                             ##
  ##===========================================================================================##

  # Update the position:

  x += v_mid * dt

  # Update the velocity:

  v += a_mid * dt

  ##===========================================================================================##
  ## Update the State:                                                                         ##
  ##===========================================================================================##

  # Update the local state position:

  state_local['Position'] = x

  # Update the local state velocity:

  state_local['Velocity'] = v

  # Update the local state time:

  state_local['Time'] += dt

  ##===========================================================================================##
  ## Return the Updated State:                                                                 ##
  ##===========================================================================================##

  return state_local

##=============================================================================================##
## Define the 1D Runge-Kutta Change Function:                                                  ##
##=============================================================================================##

def cf_runge_kutta_1d(state, parameters, acceleration):

  # Create a local copy of the state object:

  state_local = state.copy()

  # Unpack the state:

  y = state_local['Position'].copy()

  v = state_local['Velocity'].copy()

  # Unpack the parameters:

  dt = parameters['dt']

  # Compute the acceleration:

  a = acceleration(state_local, parameters)

  ##===========================================================================================##
  ## Calculate the k1 Values (Slope at the Beginning of the Time Step):                        ##
  ##===========================================================================================##

  k1_y = v

  k1_v = a

  ##===========================================================================================##
  ## Calculate the k2 Values (Slope at the Midpoint of the Time Step, Based on k1 Values):     ##
  ##===========================================================================================##

  # Position k2 values:

  k2_y = v + 0.5 * k1_v * dt

  # Velocity k2 values:

  y_mid_1 = y + 0.5 * k1_y * dt

  v_mid_1 = v + 0.5 * k1_v * dt

  # Acceleration k2 values:

  state_local['Position'] = y_mid_1

  state_local['Velocity'] = v_mid_1

  k2_v = acceleration(state_local, parameters)

  ##===========================================================================================##
  ## Calculate the k3 Values (Slope at the Midpoint of the Time Step, Based on k2 Values):     ##
  ##===========================================================================================##

  # Position k3 values:

  k3_y = v + 0.5 * k2_v * dt

  # Velocity k3 values:

  y_mid_2 = y + 0.5 * k2_y * dt

  v_mid_2 = v + 0.5 * k2_v * dt

  # Acceleration k3 values:

  state_local['Position'] = y_mid_2

  state_local['Velocity'] = v_mid_2

  k3_v = acceleration(state_local, parameters)

  ##===========================================================================================##
  ## Calculate the k4 Values (Slope at the End of the Time Step, Based on k3 Values):          ##
  ##===========================================================================================##

  # Position k4 values:

  k4_y = v + k3_v * dt

  # Velocity k4 values:

  y_end = y + k3_y * dt

  v_end = v + k3_v * dt

  # Acceleration k4 values:

  state_local['Position'] = y_end

  state_local['Velocity'] = v_end

  k4_v = acceleration(state_local, parameters)

  ##===========================================================================================##
  ## Update the State Values:                                                                  ##
  ##===========================================================================================##

  # Update the object's velocity:

  v += (k1_v + 2.0 * k2_v + 2.0 * k3_v + k4_v) * dt / 6.0

  # Update the object's position:

  y += (k1_y + 2.0 * k2_y + 2.0 * k3_y + k4_y) * dt / 6.0

  # Update the local state positions:

  state_local['Position'] = y

  # Update the local state velocity:

  state_local['Velocity'] = v

  # Update the local state time:

  state_local['Time'] += dt

  # Return the updated local state:

  return state_local
```

### <b> Define the Run Simulation and Analyze Simulation Functions </b>

```python
#@title This cell defines the run_simulation and analyze_simulation functions.

##=============================================================================================##
## Create a Function to Set Up and Run the Simulation:                                         ##
##=============================================================================================##

def run_simulation(parameters, initial_state, change_function, acceleration, show_wtime = False):

  ##===========================================================================================##
  ## Get the Simulation Start Time:                                                            ##
  ##===========================================================================================##

  wall_start_time = tm.time()

  ##===========================================================================================##
  ## Unpack the Simulation Parameters:                                                         ##
  ##===========================================================================================##

  # Get the end time:

  end_time = parameters.loc['t_end']

  ##===========================================================================================##
  ## Create the System's State History:                                                        ##
  ##===========================================================================================##

  history = [initial_state.copy()]

  indices = [0]

  ##===========================================================================================##
  ## Run the Simulation for the Desired Number of Time Steps:                                  ##
  ##===========================================================================================##

  end_condition = False

  while (end_condition == False):

    # Get the current loop iteration:

    index = len(history)

    indices.append(index)

    # Get the current system state:

    state_old = history[index - 1]

    # Update the state of the system for the current time step:

    state_new = change_function(state_old.copy(deep = True), parameters, acceleration)

    # Store the new system state in the state_history DataFrame:

    history.append(state_new)

    # If an end condition has been reached, break the loop:

    if (state_new['Time'] >= end_time): end_condition = True

  ##===========================================================================================##
  ## Create the System's State History:                                                        ##
  ##===========================================================================================##

  # Create the state_history DataFrame:

  state_history = pd.DataFrame(history)

  state_history.name = 'Motion Over Time'

  state_history.index = indices

  state_history.index.name = 'Time Step'

  ##===========================================================================================##
  ## Get the Simulation End Time:                                                              ##
  ##===========================================================================================##

  wall_end_time = tm.time()

  if (show_wtime == True):

    print("Elapsed Simulation Runtime:", np.round(wall_end_time - wall_start_time, 5), "seconds")

  ##===========================================================================================##
  ## Return the System's State History:                                                        ##
  ##===========================================================================================##

  return state_history

##=============================================================================================##
## Create a Function to Analyze the Simulation Results:                                        ##
##=============================================================================================##

def analyze_simulation(state_history, parameters, loss_function, acceleration, x_list = [],
                       y_list = [], color_list = [], title_list = [], x_label_list = [],
                       y_label_list = [], aspect_list = [], plot = False, display = False):

  ##===========================================================================================##
  ## Compute the Simulation_Loss:                                                              ##
  ##===========================================================================================##

  loss_history = loss_function(parameters, state_history, acceleration)

  ##===========================================================================================##
  ## Add the Loss History to the State History:                                                ##
  ##===========================================================================================##

  state_history['Loss'] = loss_history

  ##===========================================================================================##
  ## If requested, display the history data:                                                   ##
  ##===========================================================================================##

  if (display == True):

    display_dataframes([state_history, state_history.iloc[-1]],["State History", "Final State"],
                       n_items = 5)

  ##===========================================================================================##
  ## If requested, plot the history data:                                                      ##
  ##===========================================================================================##

  if (plot == True):

    plot_data([state_history], x_list, y_list, color_list, title_list, x_label_list,
              y_label_list, aspect_list, legend_list = [])

  ##=============================================================================================##
## Create a Function to Analyze the Simulation Results:                                        ##
##=============================================================================================##

def analyze_simulation2(state_history, parameters, loss_function, acceleration, x_list = [],
                       y_list = [], color_list = [], title_list = [], x_label_list = [],
                       y_label_list = [], aspect_list = [], plot = False, display = False):

  ##===========================================================================================##
  ## Compute the Simulation_Loss:                                                              ##
  ##===========================================================================================##

  loss_history = loss_function(parameters, state_history, acceleration)

  ##===========================================================================================##
  ## Add the Loss History to the State History:                                                ##
  ##===========================================================================================##

  state_history['Loss'] = loss_history

  ##===========================================================================================##
  ## If requested, display the history data:                                                   ##
  ##===========================================================================================##

  if (display == True):

    display_dataframes([state_history, state_history.iloc[-1]],["State History", "Final State"],
                       n_items = 5)

  ##===========================================================================================##
  ## If requested, plot the history data:                                                      ##
  ##===========================================================================================##

  if (plot == True):

    state_history_1 = state_history.copy()

    state_history_1['Position'] = state_history_1['Position'].apply(lambda x: x[0])

    state_history_1['Velocity'] = state_history_1['Velocity'].apply(lambda x: x[0])

    state_history_2 = state_history.copy()

    state_history_2['Position'] = state_history_2['Position'].apply(lambda x: x[1])

    state_history_2['Velocity'] = state_history_2['Velocity'].apply(lambda x: x[1])

    plot_data([state_history_1], x_list, y_list, color_list, title_list, x_label_list,
              y_label_list, aspect_list, legend_list = [])

    plot_data([state_history_2], x_list, y_list, color_list, title_list, x_label_list,
              y_label_list, aspect_list, legend_list = [])
```

### <b> Define the Coupled Spring Oscillator Simulation Parameters </b>

```python
##=============================================================================================##
## Simple Harmonic Oscillator Spring Parameters:                                               ##
##=============================================================================================##

# Equilibrium Length (meters):

L_eq = 1.0

# Spring Force Coefficient (N / m):

k = np.pi**2

# Time step size (in seconds):

dt = 0.01

# Simulation end time (in seconds):

t_end = 10

##=============================================================================================##
## Store the parameters in a Pandas Series:                                                    ##
##=============================================================================================##

parameters_spring_coupled = pd.Series({
    "L_eq":   L_eq,
    "k":      k,
    "dt":     dt,
    "t_end":  t_end
})

##=============================================================================================##
## Rename the Columns:                                                                         ##
##=============================================================================================##

parameters_spring_coupled.name = 'Value'

parameters_spring_coupled.index.name = "Parameter"
```

### <b> Define the Coupled Spring Oscillator Initial State </b>

```python
##=============================================================================================##
## Create the State Object                                                                     ##
##=============================================================================================##

# Initial Position (meters):

x_initial = np.array([-0.5, 0.5])

# Initial Velocity (meters per second):

v_initial = np.array([0.0, 0.0])

# Initial Time (seconds):

t_initial = 0

##=============================================================================================##
## Store the initial state in a Pandas Series:                                                 ##
##=============================================================================================##

initial_state_spring_coupled = pd.Series({
    "Position" : x_initial,
    "Velocity" : v_initial,
    "Time"     : t_initial
})

initial_state_spring_coupled.name = 'Value'

initial_state_spring_coupled.index.name = "State"

##=============================================================================================##
## Display the Parameters and Initial State:                                                   ##
##=============================================================================================##

display_dataframes([parameters_spring_coupled, initial_state_spring_coupled],
                   ["Coupled Spring Parameters", "Coupled Spring Initial State"], n_items = 15)
```

##<font size = 5> <b> 2.3 Define the Coupled Spring Oscillator Acceleration Function </b> </font>

```python
##=============================================================================================##
## Define the Acceleration Function For the Falling Object Simulation:                         ##
##=============================================================================================##

def acceleration_spring_coupled(state, parameters):

  ##===========================================================================================##
  ## Unpack the state and parameters:                                                          ##
  ##===========================================================================================##

  # Get the unstretched spring length:

  L_eq = parameters["L_eq"]

  # Get the spring force constant:

  k = parameters["k"]

  # Get the objects' current positions:

  x = state["Position"].copy()

  # Get the objects' current velocities:

  v = state["Velocity"].copy()

  ##===========================================================================================##
  ## Calculate the Acceleration of Each Object:                                                ##
  ##===========================================================================================##

  # Define the stretch of spring 1 (left-most):

  stretch_1 = x[0] + 0.5 * L_eq

  # Define the stretch of spring 2 (middle):

  stretch_2 = x[1] - x[0] - L_eq

  # Define the stretch of spring 3 (right-most):

  stretch_3 = 0.5 * L_eq - x[1]

  # Calculate the acceleration of object 1:

  a1 = -k * (stretch_1) + k * (stretch_2)

  # Calculate the acceleration of object 2:

  a2 =  k * (stretch_3) - k * (stretch_2)

  # Package the acceleration for return:

  acceleration = np.array([a1, a2])

  ##===========================================================================================##
  ## Return the Acceleration:                                                                  ##
  ##===========================================================================================##

  return acceleration.copy()
```

##<font size = 5> <b> 2.4 Define the Coupled Spring Oscillator Loss Function </b> </font>

```python
##=============================================================================================##
## Create a Function to Calculate Loss:                                                        ##
##=============================================================================================##

def loss_spring_coupled(parameters, state_history, acceleration):

  ##===========================================================================================##
  ## Unpack Parameter Values:                                                                  ##
  ##===========================================================================================##

  # Get the time step:

  dt = parameters['dt']

  # Get the pendulum length:

  L_eq = parameters['L_eq']

  # Get the coupling spring constant:

  k = parameters['k']

  ##===========================================================================================##
  ## Unpack History Values:                                                                    ##
  ##===========================================================================================##

  # Get the velocity values:

  v_1_history = state_history['Velocity'].apply(lambda x: x[0]).copy()

  v_2_history = state_history['Velocity'].apply(lambda x: x[1]).copy()

  v_1_initial = v_1_history[0]

  v_2_initial = v_2_history[0]

  # Get the position values:

  x_1_history = state_history['Position'].apply(lambda x: x[0]).copy()

  x_2_history = state_history['Position'].apply(lambda x: x[1]).copy()

  x_1_initial = x_1_history[0]

  x_2_initial = x_2_history[0]

  # Get the initial spring stretch values:

  stretch_1_initial = x_1_initial + 0.5 * L_eq

  stretch_2_initial = x_2_initial - x_1_initial - L_eq

  stretch_3_initial = 0.5 * L_eq - x_2_initial

  # Get the history spring stretch values:

  stretch_1_history = x_1_history + 0.5 * L_eq

  stretch_2_history = x_2_history - x_1_history - L_eq

  stretch_3_history = 0.5 * L_eq - x_2_history

  ##===========================================================================================##
  ## Calculate the Total Energy Initial:                                                       ##
  ##===========================================================================================##

  potential_energy_initial = \
    0.5 * k * ((stretch_1_initial)**2 + (stretch_2_initial)**2 + (stretch_3_initial)**2)

  kinetic_energy_initial = 0.5 * (v_1_initial**2 + v_2_initial**2)

  total_energy_initial = potential_energy_initial + kinetic_energy_initial

  ##===========================================================================================##
  ## Calculate the Total Energy History:                                                       ##
  ##===========================================================================================##

  potential_energy_history =  \
    0.5 * k * ((stretch_1_history)**2 + (stretch_2_history)**2 + (stretch_3_history)**2)

  kinetic_energy_history = 0.5 * (v_1_history**2 + v_2_history**2)

  total_energy_history = potential_energy_history + kinetic_energy_history

  ##===========================================================================================##
  ## Calculate the Loss History:                                                               ##
  ##===========================================================================================##

  loss_history = (total_energy_history - total_energy_initial) / total_energy_initial

  ##===========================================================================================##
  ## Return the Loss History:                                                                  ##
  ##===========================================================================================##

  return loss_history
```

##<font size = 5> <b> 2.5 Run the Coupled Spring Oscillator Simulation </b> </font>

```python
##=============================================================================================##
## Run the Simulation and Get the Sate History:                                                ##
##=============================================================================================##

# Set the desired parameter set:

parameters = parameters_spring_coupled.copy()

parameters['t_end'] = 10

# Set the desired initial state:

state = initial_state_spring_coupled.copy()

#state['Position'] = np.array([-0.5, 0.5])

#state['Position'] = np.array([-1.0, 1.0])

#state['Position'] = np.array([-0.25, 0.75])

state['Position'] = np.array([-0.15, 0.5])

# Set the desired acceleration function:

acceleration = acceleration_spring_coupled

# Set the desired change function:

change_function = cf_euler_richardson_1d

# Run the simulation:

history_spring_coupled = run_simulation(parameters, state, change_function, acceleration)

##=============================================================================================##
## Analyze the Sate History and Get the Simulion Loss:                                         ##
##=============================================================================================##

# Set the history to be plotted:

history = history_spring_coupled

# Set the Loss Function:

loss = loss_spring_coupled

# Set the plotting parameters:

y_list   = ['Position', 'Velocity']
x_list   = ['Time', 'Time']
colors   = ['blue', 'red']
titles   = ['Position Vs Time', 'Velocity Vs Time']
y_labels = ['Angle (rad)', 'Velocity (rad/s)']
x_labels = ['Time (s)', 'Time (s)']
aspects  = [0.6, 0.6]

# Run the analyze simulation function:

analyze_simulation2(history, parameters, loss, acceleration, x_list, y_list, colors, titles,
                   x_labels, y_labels, aspects, plot= True, display = False)
```

##<font size = 5> <b> 3.3 Define the Pendulum Acceleration Functions </b> </font>

### <b> Define the Actual Pendulum Acceleration Function </b>

```python
##=============================================================================================##
## Define the Acceleration Function For the Falling Object Simulation:                         ##
##=============================================================================================##

def acceleration_pendulum_coupled(state, parameters):

  ##===========================================================================================##
  ## Unpack the state and parameters:                                                          ##
  ##===========================================================================================##

  # Get the local acceleration of gravity:

  g = parameters["g"]

  # Get the pendulum length:

  L = parameters["length"]

  # Get the pendulum separation:

  separation = parameters["separation"]

  # Get the coupling spring constant:

  k = parameters["k"]

  # Get the object's current position:

  theta = state["Position"].copy()

  # Get the object's current velocity:

  v = state["Velocity"].copy()

  ##===========================================================================================##
  ## Calculate the Acceleration of Each Pendulum:                                              ##
  ##===========================================================================================##

  # Cartesian coordinates of each bob:

  x1 =  L * np.sin(theta[0])
  y1 = -L * np.cos(theta[0])

  x2 = separation + L * np.sin(theta[1])
  y2 = -L * np.cos(theta[1])

  # Vector from bob1 to bob2:

  dx = x2 - x1
  dy = y2 - y1

  distance = np.sqrt(dx**2 + dy**2)

  # Spring extension (positive if stretched)

  stretch = distance - separation

  # Spring force magnitude

  F_spring = k * stretch

  # Components of spring force on each bob (along the spring direction)

  Fx = F_spring * (dx / distance)
  Fy = F_spring * (dy / distance)

  # Forces on pendulum 1 and 2 (equal and opposite):

  # force *on 1* from 2:

  Fx1 = Fx
  Fy1 = Fy

  # force *on 2* from 1

  Fx2 = -Fx
  Fy2 = -Fy

  # Torque about the pivot for each pendulum:
  # Torque = -m*g*L*sin(theta) + (Fx*cos(theta) - Fy*sin(theta))*L
  # => acc = Torque / (m*L^2)

  # Calculate acceleration_1:

  a1 = (-g * np.sin(theta[0]) + (Fx1 * np.cos(theta[0]) - Fy1 * np.sin(theta[0]))) / L

  # Calculate acceleration_2:

  a2 = (-g * np.sin(theta[1]) + (Fx2 * np.cos(theta[1]) - Fy2 * np.sin(theta[1]))) / L

  # Calculate the acceleration:

  acceleration = np.array([a1, a2])

  ##===========================================================================================##
  ## Return the Acceleration:                                                                  ##
  ##===========================================================================================##

  return acceleration.copy()
```

##<font size = 5> <b> 3.4 Define the Pendulum Loss Function </b> </font>

```python
##=============================================================================================##
## Create a Function to Calculate Loss:                                                        ##
##=============================================================================================##

def loss_pendulum_coupled(parameters, state_history, acceleration):

  ##===========================================================================================##
  ## Unpack Parameter Values:                                                                  ##
  ##===========================================================================================##

  # Get the time step:

  dt = parameters['dt']

  # Get the pendulum length:

  L = parameters['length']

  # Get the coupling spring constant:

  k = parameters['k']

  # Get the local acceleration due to gravity:

  g = parameters['g']

  ##===========================================================================================##
  ## Unpack History Values:                                                                    ##
  ##===========================================================================================##

  # Get the velocity values:

  v_1_history = state_history['Velocity'].apply(lambda x: x[0]).copy()

  v_2_history = state_history['Velocity'].apply(lambda x: x[1]).copy()

  v_1_initial = v_1_history[0]

  v_2_initial = v_2_history[0]

  # Get the position values:

  theta_1_history = state_history['Position'].apply(lambda x: x[0]).copy()

  theta_2_history = state_history['Position'].apply(lambda x: x[1]).copy()

  theta_1_initial = theta_1_history[0]

  theta_2_initial = theta_2_history[0]

  # Get the spring stretch values:

  stretch_initial = L * (np.sin(theta_2_initial) - np.sin(theta_1_initial))

  stretch_history = L * (np.sin(theta_2_history) - np.sin(theta_1_history))

  ##===========================================================================================##
  ## Calculate the Total Energy Initial:                                                       ##
  ##===========================================================================================##

  potential_energy_initial = g * L * (2 - np.cos(theta_1_initial) - np.cos(theta_2_initial)) + \
                             0.5 * k * (stretch_initial)**2

  kinetic_energy_initial = 0.5 * L**2 * (v_1_initial**2 + v_2_initial**2)

  total_energy_initial = potential_energy_initial + kinetic_energy_initial

  ##===========================================================================================##
  ## Calculate the Total Energy History:                                                       ##
  ##===========================================================================================##

  potential_energy_history = g * L * (2 - np.cos(theta_1_history) - np.cos(theta_2_history)) + \
                             0.5 * k * (stretch_history) ** 2

  kinetic_energy_history = 0.5 * L**2 * (v_1_history**2 + v_2_history**2)

  total_energy_history = potential_energy_history + kinetic_energy_history

  ##===========================================================================================##
  ## Calculate the Loss History:                                                               ##
  ##===========================================================================================##

  loss_history = (total_energy_history - total_energy_initial) / total_energy_initial

  ##===========================================================================================##
  ## Return the Loss History:                                                                  ##
  ##===========================================================================================##

  return loss_history
```

##<font size = 5> <b> 4.1 What is a Fourier Transform? </b> </font>

Let's create a signal composed of two sine waves with different frequencies and use a Fourier Transform to pick out those frequencies.

```python
##=============================================================================================##
## Generate a Sample Signal Composed of Two Sine Waves With Different Frequencies:             ##
##=============================================================================================##

# Set the sampling rate (data points per second):

sampling_rate = 1000

# Set the duration of the signal (seconds):

duration = 1.0

# Create an array of time data points:

time = np.linspace(0, duration, int(sampling_rate * duration))

# Set the frequency of sine wave 1 (in Hz):

frequency_1 = 10

# Set the amplitude of sine wave 1:

amplitude_1 = 0.7

# Set the frequency of sine wave 2 (in Hz):

frequency_2 = 25

# Set the amplitude of sine wave 2:

amplitude_2 = 0.3

# Construct sample signal 1:

signal_1 = amplitude_1 * np.sin(2 * np.pi * frequency_1 * time)

# Construct sample signal 2:

signal_2 = amplitude_2 * np.sin(2 * np.pi * frequency_2 * time)

# Construct the sample signal:

signal = signal_1 + signal_2

##=============================================================================================##
## Perform the Fourier Transform and Extract the Frequency Data:                               ##
##=============================================================================================##

# Get the length of the sample signal (number of data points):

N = len(signal)

# Use numpy.fft (fast Fourier Transform)

yf = np.fft.fft(signal)

xf = np.fft.fftfreq(N, 1 / sampling_rate)

##=============================================================================================##
## Process and the Results:                                                                    ##
##=============================================================================================##

# The FFT output is symmetric for real-values, so we typically plot only positive frequencies.
# We also take the absolute value to get the magnitude (amplitude) of each frequency component.

n_oneside = N // 2

f_oneside = xf[:n_oneside]

# Normalize the amplitude:

X_oneside = np.abs(yf[:n_oneside]) / n_oneside

##=============================================================================================##
## Create the Plots:                                                                           ##
##=============================================================================================##

fig, axs = plt.subplots(2, 2, figsize=(10, 8))

# Signal 1 Plot:

axs[0, 0].plot(time, signal_1)
axs[0, 0].set_title('Signal 1')
axs[0, 0].set_xlabel('Time (s)')
axs[0, 0].set_ylabel('Amplitude')
axs[0, 0].grid(True)

# Signal 2 Plot:

axs[0, 1].plot(time, signal_2)
axs[0, 1].set_title('Signal 2')
axs[0, 1].set_xlabel('Time (s)')
axs[0, 1].set_ylabel('Amplitude')
axs[0, 1].grid(True)

# Combined Signal Plot:

axs[1, 0].plot(time, signal)
axs[1, 0].set_title('Combined Signal')
axs[1, 0].set_xlabel('Time (s)')
axs[1, 0].set_ylabel('Amplitude')
axs[1, 0].grid(True)

# Fourier Transform Plot:

axs[1, 1].plot(f_oneside, X_oneside)
axs[1, 1].set_title('FFT Amplitude Spectrum')
axs[1, 1].set_xlabel('Frequency (Hz)')
axs[1, 1].set_ylabel('Normalized Amplitude')
axs[1, 1].set_xlim(0, 40)
axs[1, 1].grid(True)

plt.tight_layout()
plt.show()
```
