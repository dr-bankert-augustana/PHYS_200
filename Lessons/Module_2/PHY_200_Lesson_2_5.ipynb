##<font size = 5> <b> 2.1 Linear Regression </b> </font>

In this lesson, we will learn the basic structure of neural networks as a
generalization of linear regression.

<br>

Let's recap our setup. We start with a dataset that consists of measurements from some system. We have (D) <b>features</b> (independent variables), (x$_1$ through x$_D$), and one <b>target</b> (dependent variable) (y). Each of these will be arranged as aa seperate column in out DataFrame.

<br>

We have (N) samples (data points) for (x) and (y), which we arrange into the rows of our DataFrame.

---

<br>

## <b>Expanded Matrix Form of Linear Model:</b>

$$
\hat{y}\of{x} = A^T \phi(x) + b
$$

where:

- $\hat{y}\of{x}$ is our predictions of the features
- $A$ are the model weights (linear coefficients)
- $A^T$ is the transpose of $A$
- $b$ are the model biases (y-intercepts)  

<br>

One thing that we've found useful is to define a set of features, ($\phi_0$ through $\phi_{m-1}$), based on the original inputs (x). These are nonlinear transformations of (x) that are designed to improve the performance of the model. We've seen examples of this in linear regression where nonlinear features introduced curvature into our regression lines.

<br>

Recall that in linear regression, we proposed a simple model, $\hat{y}\of{x}$, that predicted the output as a linear combination of the inputs. In the one-dimensional case, this is simply a line with intercept (b) and slope
(a). The task of linear regression was to find the values of (a) and (b) that
minimized the loss function.

<br>

In general, if we have 𝑚 features instead of just one, then the model is a
sum of 𝑚 terms. Those are b + ($a_1 \spc \phi_1\of{x}$ through $a_{m-1} \spc \phi_{m-1}\of{x}$. We can express this compactly in a vector notation by defining a weight vector ($A$), a feature vector ($\phi\of{x}$), and an offset ($b$). The model then becomes $\hat{y}\of{x} = A^T \spc \phi\of{x} + b$. The offset is sometimes called the bias, but this really has nothing to do with the statistical bias of the model. So don't let that confuse you.

<br>

---

## <b>Example: Two Input Features</b>

<br>

###<b>Setup:</b>

Suppose our model has two features ($x_1$ and $x_2$)

$$
\phi(x) =
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
$$

<br>

Next we define our bias terms for $\hat{y}$ (value of y when all features are zero):

$$
\text{bias} = b
$$

<br>

Now define our model weights (the linear coefficients for $x_1$ and $x_2$):

$$
A =
\begin{bmatrix}
w_1 \\
w_2
\end{bmatrix}
$$

Then:

$$
A^T =
\begin{bmatrix}
w_1 & w_2
\end{bmatrix}
$$

<br>

### <b>Matrix Multiplication Step:</b>

Now compute:

$$
\hat{y}\of{x} = A^T \phi(x) = \
\begin{bmatrix}w_1 & w_2
\end{bmatrix}
\begin{bmatrix}
x_1 \\
x_2
\end{bmatrix}
$$

Carrying out the multiplication:

$$
\hat{y}\of{x} = w_1 x_1 + w_2 x_2 + b
$$

---

### Geometric Interpretation

The equation

$$
w_1 x_1 + w_2 x_2 + b = 0
$$

defines a **line in 2D feature space**.

- The weight vector determines the orientation.
- The bias shifts the line.

In higher dimensions, this becomes a **hyperplane**.

<br>

<b>Our goal is to construct a model that, given some previously unseen input value (x$_{\text{new}}$), will make a good prediction for its corresponding target value (y).</b>

---

The linear regression problem can then be expressed very succinctly as
finding the parameters (a) and (b) that minimize the loss function, which is the mean square of the difference between the predicted output ($h\of{x_i}$) and the true output (y_i).


<center>
<font size = 5>
$
\begin{array}{l l}
\textrm{RMSE} & = \sqrt{\displaystyle \frac{1}{N} \displaystyle\sum\limits_{i=1}^{N} \hspace{0.2 pc} (e_i)^2}\\&\\
& = \sqrt{\displaystyle \frac{1}{N} \displaystyle\sum\limits_{i=1}^{N} \hspace{0.2 pc} (y_i - A^T \spc \phi\of{x_i} - b)^2}
\end{array}
$
</font>
</center>

##<font size = 5> <b> 2.2 Extension to Neural Networks </b> </font>

Neural networks start from the same picture and generalize it in several
ways. First, they use very specific feature functions, which in the neural
network community are called activation functions. A few of the most
common activation functions are:

* In a neural network, it acts like a switch, either opening or closing a channel.

* The tanh or hyperbolic tangent function; this function puts saturation limits on the output of the neuron. It allows small values to go through more or less unchanged, but truncates large positive or negative values.

* The rectified linear unit or ReLU. This just takes the maximum between the input value and 0. And thus it prevents any negative numbers from going through.

<br>

A second generalization of neural networks is that they add coefficients to
all of the lines in the graph, not just the ones on the output side. So whereas
before with linear regression, the input to the feature vector was simply 𝑥.
Now with neural networks, it will be a matrix $𝐴_1 \spc 𝑥$ plus an offset vector $𝑏_1$.

<br>

The third generalization of neural networks is that they allow for many
layers to be used. These layers are set out sequentially so that each feature
vector $\phi$ is fed by a matrix $𝐴$ multiplied by the output of the previous layer plus an offset vector $𝑏$. Let's use capital letter $𝐿$ for the number of layers in the network and count them from left to right. There are $𝐿 − 1$ internal or hidden layers. And the last one is the output layer. We'll use lowercase $𝑐$ for the outputs from each layer. The final output of the model $ℎ\of{x}$ is the transformation of $𝑐_{𝐿−1}$ with coefficients $𝐴_𝐿$ and $𝑏_𝐿$. Then $𝑐_{𝐿−1}$ is the output of layer $𝐿 − 1$. This equals the feature $\phi_{𝐿−1}$ applied to the input of that layer, which is the transformation of the output of the previous layer through coefficients $𝐴_{𝐿−1}$ and $𝑏_{𝐿−1}$. And we continue in this way, feeding each layer
with the output of the previous layer, which is transformed through a dense
mesh of coefficients $𝐴$ and $𝑏$ until we reach the input.

<br>

A final generalization has to do with the output layer. By adding a sigmoid
activation function to the output node, we immediately transform the regression model into a binary classification model. Later we will see other
output layers that can be used to create multiclass classification models.

##<font size = 5> <b> 2.3 A Very Simple Neural Network In Action </b> </font>

### <b> Import Libraries and Define Useful Functions</b>

```python
##=============================================================================================##
## Import Libraries:                                                                           ##
##=============================================================================================##

import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from IPython.display import display_html
```

```python
#@title This cell defines the functions: display_dataframes and plot_data

##=============================================================================================##
## Function:  display_dataframes                                                               ##
##                                                                                             ##
## Purpose:   Display multiple DataFrames side-by-side with titles                             ##
##                                                                                             ##
## Input(s):  dataframe_list - List of DataFrames to be displayed                              ##
##            title_list     - List of titles for the displayed DataFrames                     ##
##            n_items        - Number of items to display (optional, default = 5)              ##
##                                                                                             ##
## Output(s): None                                                                             ##
##=============================================================================================##

def display_dataframes(dataframe_list, title_list, n_items = 5, tail = False):

  ##===========================================================================================##
  ## Create a String for Housing the Commands to Be Sent to the display_html() Function:       ##
  ##===========================================================================================##

  html_str = ""

  ##===========================================================================================##
  ## Loop Over the Elements in the item_list and title_list:                                   ##
  ##===========================================================================================##

  for df, title in zip(dataframe_list, title_list):

    # Convert the current dataframe info to html:

    if (tail == True):

      html_df = pd.DataFrame(df).tail(n_items).to_html()

    else:

      html_df = pd.DataFrame(df).head(n_items).to_html()

    ##=========================================================================================##
    ## Wrap title and DataFrames in a Styled HTML <div>:                                       ##
    ##=========================================================================================##

    html_str += "<div style='display: inline-block; margin-right: 20px; vertical-align: top;'>"

    html_str += "<h3 style='text-align: center;'>" + str(title) + "</h3><hr>" + str(html_df)

    html_str += "</div>"

  ##===========================================================================================##
  ## Send the HTML String to the display_html() Function:                                      ##
  ##===========================================================================================##

  display_html(html_str, raw = True)
```

### <b> Load The Data </b>

Below, we load in the titanic dataset from seaborn. Note that this dataset has missing values in the age column that are subsequently dropped. In this exercise we will only use the top five rows as that dataset in order to clarify the initial workings of the Artificial Neural Network.

```python
##=============================================================================================##
## Load Titanic Passenger Data and Parse the Features and Target Data:                         ##
##=============================================================================================##

# Load Titanic data:

titanic = sns.load_dataset('titanic').dropna(subset = ['age'])

# Set the feature data (passenger age and ticket price):

X = titanic[['age', 'fare']].head()

# Set the target data (1 = survived, 0 = did not survive):

y = titanic['survived'].head()

# Display the feature and target data:

display_dataframes([X, y], ["Features (X)", "Target (y)"])
```

### <b> Add Bias Terms </b>

The first step is to add a column of ones to act as the bias feature for the data. As a convention, we will add the bias column as the first column in the dataset. Below, we create a function that adds a bias feature as the first column to the input array X.

```python
##=============================================================================================##
## Define a Function to Add a Bias Column to A Features DataFrame:                             ##
##=============================================================================================##

def add_bias(X):

  # Create the Bias Data:

  bias = pd.Series(np.ones(len(X)), name = 'bias')

  # Add the Bias Data to the Feature Data:

  X_new = pd.concat([bias, X], axis = 1)

  return X_new
```

```python
##=============================================================================================##
## Test the add_bias Function:                                                                 ##
##=============================================================================================##

X_with_bias = add_bias(X)

display_dataframes([X_with_bias, y], ["Features With Bias (X)", "Target (y)"])
```

### <b> Create the Initial Weights </b>

The first step in building the Neural Network is to assign a weight for each input for each node. If we use a single node, this amounts to creating three initial weights -- one for the bias and one weight for each feature. Below, we create such weights using NumPy and its random normal random number generator. Below, we generate an array of (3) weights for the bias and features.

```python
##=============================================================================================##
## Create a Random Set of Initial Weights:                                                     ##
##=============================================================================================##

weights = np.random.normal(size = 3)

print("Initial Weights: ", weights)
```

### <b> Computing the Weighted Sum </b>

Next, we use the weights to apply them to the features and bias. This can be done using matrix multiplication or a loop.

<br>

<b>Matrix multiplication is preferred</b>.

```python
##=============================================================================================##
## Perform the Matrix Multiplication of the Biased Features and Weights:                       ##
##=============================================================================================##

weighted_sum = X_with_bias@weights

display_dataframes([pd.DataFrame(weighted_sum)], ["Weighted Sum"])
```

### <b> Sigmoid Activation </b>

After computing the weighted sum for the data we apply a sigmoid activation function.  Below, we create the `sigmoid` function to take in an array of values and return the result of applying the transformation:

$$f(x) = \frac{1}{1 + e^{-x}}$$


Interpretation of the sigmoid output:

Values near 1 → likely survived

Values near 0 → unlikely survived

```python
##=============================================================================================##
## Define a Function to Perform a Sigmoid Activation:                                          ##
##=============================================================================================##

def sigmoid(X):

    return 1/(1 + np.exp(-X))
```

```python
##=============================================================================================##
## Apply the Sigmoid Activation Function to the Weighted Sum:                                  ##
##=============================================================================================##

output = sigmoid(weighted_sum)

display_dataframes([pd.DataFrame(output)], ["Output"])
```

### <b> Making Predictions </b>

The output of the sigmoid can be interpreted as a probability of being a member of the positive class -- survived. Next, we use the output to make predictions for our first pass through the neural network.

```python
##=============================================================================================##
## Make Prediction By: output > 0.5 -> survived (1), output <= 0.5 -> did not survive (0)      ##
##=============================================================================================##

# Get our predictions:

preds = np.where(output > 0.5, 1, 0)

# Compute the accuracy of our predictions:

starting_acc = sum(preds == y)/len(y)

display_dataframes([pd.DataFrame(preds), y], ["Predictions", "Target"])
print()
print("Starting Accuracy: ", starting_acc * 100, "%")
```

Great job, you just created your first neural network! Next we try a more complex Neural Network.

##<font size = 5> <b> 2.2 Multiple Nodes </b> </font>

Now, we focuse on expanding the neural network to use three nodes for a prediction. The main difference from our prior activity is that for each new node you will need a corresponding set of weights for features and bias. Again, we use a small sample of the titanic dataset to test the networks forward pass.

```python
##=============================================================================================##
## Review our Features and Targets:                                                            ##
##=============================================================================================##

display_dataframes([X_with_bias, y], ["Features With Bias (X)", "Target (y)"])
```

### <b> A Weight Array </b>

Now, because we are working with three layers, we need to create weights for the bias and each feature in each node. This means for a 3-node layer we need three rows of three columns of weights. We create this array below as weights using np.random.normal.

```python
##=============================================================================================##
## Create a Random Set of Initial Weights:                                                     ##
##=============================================================================================##

weights = np.random.normal(size = (3, 3))

display_dataframes([pd.DataFrame(weights)], ["Initial Weights"])
```

### <b> Weighted Sum </b>

Again, we use the weights to compute a weighted sum of the features and bias term with the weights.

```python
##=============================================================================================##
## Perform the Weighted Sum:                                                                   ##
##=============================================================================================##

layer_1_output = X_with_bias@weights

display_dataframes([pd.DataFrame(layer_1_output)], ["layer_1_output"])
```

### <b> Activation Function </b>

Again, we use a sigmoid activation function to transform the results of the weighted sum using the sigmoid function to apply the activation function.

```python
##=============================================================================================##
## Use the Sigmoid Activation Function:                                                        ##
##=============================================================================================##

layer_1_activation = sigmoid(layer_1_output)

display_dataframes([pd.DataFrame(layer_1_activation)], ["layer_1_activation"])
```

### <b> Bias Term </b>

The array from above still needs to again have weights applied, and these include a weight for a bias term. We use our function to again add a bias term and assign the resulting array to layer_1_with_bias. Then we create an appropriately sized array of weights for a single node named weights_2

```python
##=============================================================================================##
## Add Bias Term and Set the Output Node Weights:                                              ##
##=============================================================================================##

layer_1_with_bias = add_bias(layer_1_activation)

weights_2 = np.random.normal(size = (4,))

display_dataframes([pd.DataFrame(layer_1_with_bias), pd.DataFrame(weights_2)], ["Layer 1 With Bias", "weights_2"])
```

### <b> Output of First Pass </b>

Finally, use the weights and bias added data from the first layer to compute the weighted sum and pass the results through a sigmoid activation. We use these results to predict the positive class if output > 0.5 and class 0 otherwise. What is the accuracy on the dataset?

```python
preds_multi = layer_1_with_bias@weights_2
acc_first_pass = sum(y == np.where(sigmoid(preds_multi) > .5, 1, 0))/len(y)

print("Accuracy: ", acc_first_pass * 100, "%")
```

##<font size = 5> <b> 2.3 Basic Weight Update </b> </font>


Now, we focuse on updating the weights to improve the performance of a network. To demonstrate this update process, we implement a basic loss function and its derivative to update randomly initialized weights. We begin with a very basic dataset meant to replicate the logic gate problems surrounding the early ANN models.

```python
##=============================================================================================##
## Set Up Our Data:                                                                            ##
##=============================================================================================##

# Set Feature Data:

X = np.array([[0, 0], [0, 1], [1, 0], [1, 1]])

# Set Target Data:

y = np.array([0, 1, 1, 1])

# Subset no and yes

nos = X[y == 0]

yesses = X[y == 1]

##=============================================================================================##
## Display Our Data:                                                                            ##
##=============================================================================================##

# Show the Data:

display_dataframes([pd.DataFrame(X), pd.DataFrame(y)], ["Features (X)", "Target (y)"])

# Plot the Data:

plt.scatter(nos[:, 0], nos[:, 1], marker = 'x', s = 400)
plt.scatter(yesses[:, 0], yesses[:, 1], marker = '+', s = 400)
plt.title("Can we Seperate These?");
```

### <b> Preparing the Data </b>

To prepare the data for modeling, we will need to add a bias term. In this example, we prepend a column of -1's to the array X and save the new array as Xt below.

```python
##=============================================================================================##
## Add Bias to Feature Data:                                                                   ##
##=============================================================================================##

# Create the Bias Data:

b = np.ones((X.shape[0], 1)) * -1

# Add the Bias Data to the Feature Data Matrix:

Xt = np.concatenate((b, X), axis = 1)

display_dataframes([pd.DataFrame(Xt)], ["Features With Bias (Xt)"])
```

### <b> Create the Initial Weights </b>

Below, we create an array of weights for both the bias term and the two features. We, then use this weight array (w) to apply these weights to the input and bias and assign the product to weighted_sum below.

```python
##=============================================================================================##
## Create and Apply the Weights Array:                                                         ##
##=============================================================================================##

# Set the weights array:

w = np.array([-0.05, -0.02, 0.02])

# Apply the weights array to the features with bias array:

weighted_sum = Xt@w

display_dataframes([pd.DataFrame(weighted_sum)], ["Weighted Sum"])
```

### <b> Use the Activation Function And Get Predictions</b>

Now, we use the basic threshold activation function to output 1 if the input is greater than 0, otherwise label it 0. We assign the values as predictions below. After we apply this threshold function to the weighted sum, we find the accuracy of the classifier after 1 epoch.

```python
##=============================================================================================##
## Make Prediction By: weighted_sum > 0 -> prediction = 1, weighted_sum <= 0 -> prediction = 0 ##
##=============================================================================================##

# Make predictions:

preds = np.where(weighted_sum > 0, 1, 0)

# Compute the accuracy of our predictions:

starting_acc = sum(preds == y)/len(y)

display_dataframes([pd.DataFrame(preds), y], ["Predictions", "Target"])
print()
print("Starting Accuracy: ", starting_acc * 100, "%")
```

### <b> Weight Update</b>

Next, we want to update the initial weights using the following rule for gradient descent with learning rate $\alpha$:

$$w_{i + 1} = w_{i} + \alpha(y - \hat{y})x$$

We use this rule to update the initial weights with $\alpha = 0.1$ and assign to `w2` below.

```python
##=============================================================================================##
## Update Weights Array Using Gradient Descent:                                                ##
##=============================================================================================##

# Create the new set of weights

w2 = w + 0.1*(y - preds)@Xt

# Display the new weights

display_dataframes([pd.DataFrame(w), pd.DataFrame(w2)], ["Starting Weights", "Updated Weights"])

##=============================================================================================##
## Check the new predictions and accuracy:                                                     ##
##=============================================================================================##

# Apply the weights array to the features with bias array:

weighted_sum2 = Xt@w2

# Make predictions:

preds2 = np.where(weighted_sum2 > 0, 1, 0)

# Compute the accuracy of our predictions:

second_acc = sum(preds2 == y)/len(y)

display_dataframes([pd.DataFrame(preds2), y], ["Predictions", "Target"])
print()
print("Starting Accuracy: ", second_acc * 100, "%")
```

### <b> Putting it together</b>

Now, we build a model by performing 10 iterations of the weight updates. What is the final accuracy?

```python
##=============================================================================================##
## Start With the Original Weights and Update Through 10 Iterations:                           ##
##=============================================================================================##

# Set Original Weights:

w = np.array([-0.05, -0.02, 0.02])

# Create list to store accuracy:

acc_list = []

for i in range(10):

  # Perform our estimation:

  predictions = np.where(Xt@w > 0, 1, 0)

  # Get our accuracy:

  acc = sum(predictions == y)/len(y)

  acc_list.append(acc)

  # Update the weights:

  w = w + 0.1*(y - predictions)@Xt

# Show the Accuracy List:

display_dataframes([pd.DataFrame(acc_list)], ["Accuracy List"], n_items = 10)
```

Excellent! We see that the Neural Network has been trained to identify the pattern.
