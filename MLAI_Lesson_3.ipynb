```python
penguin_df = pd.read_csv("./penguins.csv")

penguin_df
```

```python
penguin_df.plot(x = 'bill_length_mm', y = 'bill_depth_mm', kind = 'scatter')
```

```python
penguin_df.plot(x = 'bill_length_mm', y = 'body_mass_g', kind = 'scatter')
```

```python
penguin_df.plot(x = 'bill_depth_mm', y = 'body_mass_g', kind = 'scatter')
```

```python
figure_1 = plt.figure()

graph_1 = figure_1.add_axes([0,1.2,1,1])
graph_2 = figure_1.add_axes([1.2,1.2,1,1])
graph_3 = figure_1.add_axes([0,0,1,1])

graph_1.set_title("Species: Adelie")
graph_2.set_title("Species: Gentoo")
graph_3.set_title("Species: Chinstrap")

graph_1.set_xlabel("Bill Depth")
graph_1.set_ylabel("Body Mass")

graph_2.set_xlabel("Bill Depth")
graph_2.set_ylabel("Body Mass")

graph_3.set_xlabel("Bill Depth")
graph_3.set_ylabel("Body Mass")

#celeb_df.plot(kind='line', y=['feet','meters'], figsize=(10, 6),grid=True, ax=graph_1)

penguin_df.query("species == 'Adelie'").plot(x = 'bill_depth_mm', y = 'body_mass_g', kind = 'scatter', ax=graph_1)
penguin_df.query("species == 'Gentoo'").plot(x = 'bill_depth_mm', y = 'body_mass_g', kind = 'scatter', ax=graph_2)
penguin_df.query("species == 'Chinstrap'").plot(x = 'bill_depth_mm', y = 'body_mass_g', kind = 'scatter', ax=graph_3)
```

```python
penguin_corr = penguin_df.corr(numeric_only = True)

penguin_corr
```

```python
figure_1 = plt.figure()

graph_1 = figure_1.add_axes([0,0,1,1])

c = graph_1.pcolor(penguin_corr, cmap='RdBu')

plt.yticks(np.arange(0.5, len(penguin_corr.index), 1), penguin_corr.index)

plt.xticks(np.arange(0.5, len(penguin_corr.columns), 1), penguin_corr.columns)

figure_1.colorbar(c, ax=graph_1)

plt.show()
```

```python
def max_corr(df_in):

    max_corr_1 = df_in[df_in < 1.0].max().idxmax()
    
    df_2 = df_in.drop(max_corr_1, axis = 1)
    
    max_corr_2 = df_2[df_2 < 1.0].max().idxmax()

    return max_corr_1, max_corr_2
```

```python
max_penguin_corr_1, max_penguin_corr_2 = max_corr(penguin_corr)

print("The max correlation is between: '" + max_penguin_corr_1 + "' and '" + max_penguin_corr_2 + "'")
```

```python
penguin_df.plot(x = 'flipper_length_mm', y = 'body_mass_g', kind = 'scatter')
```

```python
def min_corr(df_in):

    min_corr_1 = df_in.min().idxmin()
    
    df_2 = df_in.drop(min_corr_1, axis = 1)
    
    min_corr_2 = df_2.min().idxmin()

    return min_corr_1, min_corr_2
```

```python
min_penguin_corr_1, min_penguin_corr_2 = min_corr(penguin_corr)

print("The minimum correlation is between: '" + min_penguin_corr_1 + "' and '" + min_penguin_corr_2 + "'")
```
